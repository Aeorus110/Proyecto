## 1. Diccionarios: relacionar claves y valores

Un diccionario almacena pares de clave y valor. La **clave** identifica de forma única una información; el **valor** es el dato asociado. Se escribe entre llaves, con dos puntos entre clave y valor y comas entre pares. Una clave puede ser texto, número o una tupla inmutable; debe ser un valor que se pueda usar de forma estable como identificador. Una lista no puede ser clave porque es mutable.

A diferencia de una lista, un diccionario no se consulta por una posición numérica sino por su clave. Las claves no se repiten: al asignar dos veces la misma clave, el último valor sustituye al anterior. Los valores sí pueden repetirse. Desde Python moderno los diccionarios conservan el orden de inserción, pero su propósito principal sigue siendo asociar identificadores con datos, no trabajar por índice.

```python
temperaturas_por_aula = {"A-01": 21.4, "A-02": 22.0, "B-01": 20.8}

print(temperaturas_por_aula)
print(temperaturas_por_aula["A-02"])
print(len(temperaturas_por_aula))
```

### Acceso directo y acceso seguro

La forma diccionario[clave] recupera un valor cuando la clave existe. Si no existe, Python genera KeyError. Esto es útil cuando la ausencia de esa clave representa un error de datos que debe investigarse.

El método get permite una consulta opcional. Si la clave no existe, devuelve None por defecto o el valor alternativo que se indique. Get no crea la clave. Úsalo cuando sea normal que un dato pueda faltar y define qué significa el valor alternativo. No ocultes con un valor por defecto una ausencia que debería haberse detectado como error.

```python
temperaturas_por_aula = {"A-01": 21.4, "A-02": 22.0}

print(temperaturas_por_aula.get("A-01"))
print(temperaturas_por_aula.get("B-01"))
print(temperaturas_por_aula.get("B-01", "sin lectura"))
```

## 2. Añadir, actualizar y eliminar

La asignación con una clave añade un par si la clave es nueva o actualiza el valor si ya existía. El método pop elimina una clave y devuelve su valor; con un valor alternativo evita KeyError cuando la clave no existe. El método update puede incorporar varios pares, pero conviene usarlo solo cuando el origen y el efecto sean claros.

Antes de actualizar una clave, decide si la operación reemplaza una medida anterior, suma información o representa un error. Un diccionario no conserva automáticamente un historial de valores de una misma clave: una nueva asignación sobrescribe la anterior. Para guardar historial habría que asociar a la clave una lista de lecturas, estructura que ya puedes interpretar gracias a las unidades previas.

```python
estado_aulas = {"A-01": "disponible", "A-02": "ocupada"}
estado_aulas["B-01"] = "disponible"
estado_aulas["A-02"] = "revisar"
estado_eliminado = estado_aulas.pop("A-01")

print("Eliminado:", estado_eliminado)
print(estado_aulas)
```

### Recorrer diccionarios

Un for sobre un diccionario recorre sus claves. Los métodos keys, values e items ofrecen vistas de claves, valores y pares respectivamente. Para producir un informe que incluya identificador y dato, items es normalmente la opción más clara porque entrega ambos valores en cada repetición.

No modifiques el tamaño de un diccionario mientras lo recorres: añadir o eliminar claves durante un for puede causar errores o resultados confusos. Si necesitas construir un resultado distinto, crea otro diccionario o una lista fuera del recorrido.

```python
temperaturas_por_aula = {"A-01": 21.4, "A-02": 22.0, "B-01": 20.8}

for aula, temperatura in temperaturas_por_aula.items():
    print(f"{aula}: {temperatura:.1f} °C")

print(list(temperaturas_por_aula.keys()))
print(list(temperaturas_por_aula.values()))
```

## 3. Contar con diccionarios

Un patrón muy útil consiste en usar una clave como categoría y el valor como contador. Para cada elemento, se consulta el contador existente; si no existe se empieza en cero; y se suma uno. Get permite expresar esa inicialización de forma breve. Este patrón aparece en análisis de texto, etiquetas, errores o resultados de clasificación.

El diccionario de conteos debe empezar vacío. Cada clave se crea en el momento en que aparece por primera vez. Al finalizar, los valores indican cuántas veces se ha observado cada categoría. El código no necesita saber de antemano todas las categorías posibles.

```python
etiquetas = ["energía", "aula", "energía", "consumo", "aula"]
conteos = {}

for etiqueta in etiquetas:
    conteos[etiqueta] = conteos.get(etiqueta, 0) + 1

print(conteos)
```

## 4. Conjuntos: pertenencia y elementos únicos

Un conjunto, de tipo set, almacena elementos únicos. Se escribe entre llaves, pero sin pares clave y valor. La repetición de un elemento no crea otra copia. Los conjuntos no tienen un orden de posiciones que debas usar: no se accede a ellos mediante índice. Un conjunto vacío se crea con set(), porque dos llaves vacías representan un diccionario vacío.

Los conjuntos son adecuados para eliminar duplicados, comprobar pertenencia y comparar grupos de elementos. Sus elementos deben ser inmutables, por lo que no se pueden añadir listas ni diccionarios. Si necesitas mostrar un conjunto de forma ordenada, convierte el resultado en una lista y ordénala explícitamente; no dependas del orden con que se imprima un set.

```python
etiquetas = ["energía", "aula", "energía", "consumo"]
etiquetas_unicas = set(etiquetas)
conjunto_vacio = set()

print(etiquetas_unicas)
print("energía" in etiquetas_unicas)
print(conjunto_vacio)
```

### Operaciones entre conjuntos

La unión reúne todos los elementos de ambos conjuntos. La intersección contiene solo los que están en ambos. La diferencia conserva los que están en el primero y no en el segundo; por ello el orden de los conjuntos importa en una diferencia. Estas operaciones se pueden expresar con métodos o con operadores de conjuntos.

Al comparar etiquetas de dos fuentes, la intersección puede indicar términos comunes, y las diferencias pueden revelar términos exclusivos. Estas operaciones describen presencia, no frecuencia: si una etiqueta aparece diez veces sigue estando una sola vez en un set.

```python
grupo_manana = {"A-01", "A-02", "B-01"}
grupo_tarde = {"A-02", "B-01", "B-02"}

print("Unión:", grupo_manana | grupo_tarde)
print("Intersección:", grupo_manana & grupo_tarde)
print("Solo mañana:", grupo_manana - grupo_tarde)
print("Solo tarde:", grupo_tarde - grupo_manana)
```

## 5. Errores frecuentes y depuración

Errores habituales: usar una clave inexistente con acceso directo; confundir un diccionario con una lista y pedir una posición; creer que una clave duplicada crea dos entradas; usar llaves vacías esperando un conjunto; e intentar indexar un set. Para depurar, muestra el tipo, las claves disponibles o el conjunto completo. Cuando necesites una salida reproducible para revisar, usa sorted sobre un conjunto de texto.

No conviertas a texto todos los datos por comodidad. Una clave numérica y el texto que representa ese número son claves diferentes. Establece una regla de normalización antes de crear diccionarios o conjuntos a partir de datos heterogéneos.

```python
aulas = {"B-01", "A-02", "A-01"}
print(type(aulas))
print(sorted(aulas))
```

Ejercicio 08.03 · Comparar etiquetas. Convierte las listas [energia, consumo, aula, eficiencia] y [aula, consumo, accesibilidad] en conjuntos. ## Proyecto 08 · Resumen de incidencias de un edificio

Construye un programa que analice estas incidencias: incidencias = [red, proyector, red, climatizacion, proyector, red, acceso]. Debe crear un diccionario con el recuento de cada tipo y un conjunto de tipos únicos. Debe mostrar un informe ordenado alfabéticamente con tipo y cantidad, indicar cuántos tipos distintos existen y comparar los tipos observados con el conjunto de tipos esperados {red, proyector, climatizacion, acceso, seguridad}.

El informe debe mostrar las incidencias esperadas que no se han observado y, si las hubiera, incidencias observadas que no están en las esperadas. Añade comentarios sobre las inicializaciones y una tabla de pruebas que incluya un tipo repetido, un tipo único y un tipo inesperado que añadas temporalmente. Explica en Markdown por qué un conjunto no sirve para conservar la frecuencia de cada incidencia.

```python
incidencias = ["red","proyector", "red", "climatizacion", "proyector", "red", "acceso"]
tipos_esperados = {"red", "proyector", "climatizacion", "acceso", "seguridad"}
recuento = {}

for incidencia in incidencias:
    recuento[incidencia] = recuento.get(incidencia, 0 ) + 1

tipos_unicos = set(recuento)

print("RECUENTO POR TIPO ORDENADO")
recuento_ordenado = dict(sorted(recuento.items()))
for tipo, cantidad in recuento_ordenado.items():
    print(f"{tipo}: {cantidad}")

print(f"NUMERO DE TIPOS UNICOS: {len(tipos_unicos)}")

incidencias_no_observadas = tipos_esperados - tipos_unicos
incidencias_no_esperadas = tipos_unicos - tipos_esperados

print(f"Incidencias no observadas: {incidencias_no_observadas}")
if incidencias_no_esperadas:
    print(f"Incidencias no esperadas: {incidencias_no_esperadas}")
else:
    print("No hay ninguna incidencia no esperada")
```

| **Caso de prueba** | **Entrada (incidencias)** | **Resultado esperado (recuento)** | **Tipos únicos (set)** | **Esperadas no observadas (esperadas - observadas)** | **Observadas no esperadas (observadas - esperadas)** |
|---|---|---|---|---|---|
| **Tipo repetido** | `["red", "red"]` | `{'red': 2}` | `{'red'}` | `{'proyector', 'climatizacion', 'acceso', 'seguridad'}` | `set()` |
| **Tipo único** | `["climatizacion"]` | `{'climatizacion': 1}` | `{'climatizacion'}` | `{'red', 'proyector', 'acceso', 'seguridad'}` | `set()` |
| **Tipo inesperado** *(Añadido temporalmente)* | `["red", "fuga_agua"]` | `{'red': 1, 'fuga_agua': 1}` | `{'red', 'fuga_agua'}` | `{'proyector', 'climatizacion', 'acceso', 'seguridad'}` | `{'fuga_agua'}` |

No sirve dado que si hay dos incidencias iguales solo se va a quedar con una
