Una clase permite representar entidades del problema combinando estado y comportamiento. No sustituye a listas, diccionarios o funciones: es una herramienta adicional para cuando varios datos y operaciones forman una unidad coherente.

## 1. Clase, objeto y atributo

Una **clase** es una plantilla que define qué información y operaciones tendrán sus objetos. Una **instancia** u objeto es un valor concreto creado a partir de esa clase. Por ejemplo, Aula puede ser la clase y A-01 una instancia. Los atributos guardan el estado propio de cada objeto; los métodos son funciones definidas dentro de la clase que actúan sobre ese estado.

No conviene crear una clase para cualquier dato aislado. Es apropiada cuando el problema contiene varias entidades del mismo tipo, cada una con varios atributos y operaciones relacionadas. Antes de diseñarla, identifica qué datos pertenecen a la entidad, qué reglas debe mantener y qué acciones ofrece.

```python
class Aula:
    def __init__(self, codigo: str, capacidad: int) -> None:
        self.codigo = codigo
        self.capacidad = capacidad

aula_a01 = Aula("A-01", 28)
print(aula_a01.codigo)
print(aula_a01.capacidad)
```

## 2. Constructor y self

El método especial __init__ se ejecuta al crear una instancia. Recibe los datos necesarios y asigna atributos mediante self. Self representa la instancia concreta que está trabajando: por eso self.codigo y self.capacidad pueden tener valores distintos en cada aula. No se pasa self al crear el objeto; Python lo proporciona automáticamente.

Las anotaciones indican que el constructor no devuelve un valor útil. Las reglas de validación deben aparecer cerca de los datos que protegen. Por ejemplo, una capacidad negativa no representa un aula válida y puede rechazarse con ValueError.

```python
class Aula:
    def __init__(self, codigo: str, capacidad: int) -> None:
        if capacidad < 0:
            raise ValueError("La capacidad no puede ser negativa.")
        self.codigo = codigo
        self.capacidad = capacidad
        self.ocupacion = 0

    def plazas_libres(self) -> int:
        return self.capacidad - self.ocupacion

    def registrar_entrada(self) -> None:
        if self.ocupacion < self.capacidad:
            self.ocupacion = self.ocupacion + 1

aula = Aula("B-02", 2)
aula.registrar_entrada()
print(aula.plazas_libres())
```

## 3. Métodos y representación

Un método que consulta el estado devuelve información sin modificarlo; un método que cambia el estado debe respetar las reglas de la entidad. Registrar una entrada no debe permitir más ocupación que capacidad. Esta protección no elimina la necesidad de probar el programa, pero concentra la regla en un lugar.

El método especial __repr__ devuelve una representación útil para depurar, mientras que __str__ se reserva para una presentación legible. Una representación clara muestra atributos que permitan identificar el objeto y su estado sin imprimir información sensible o innecesaria.

```python
class Lectura:
    def __init__(self, aula: str, temperatura: float) -> None:
        self.aula = aula
        self.temperatura = temperatura

    def __repr__(self) -> str:
        return f"Lectura(aula={self.aula!r}, temperatura={self.temperatura!r})"

lectura = Lectura("A-01", 21.4)
print(lectura)
print(repr(lectura))
```

## 4. Composición y diseño

La composición expresa que un objeto contiene o utiliza otros objetos. Por ejemplo, un Informe puede contener una lista de Lectura. Suele ser preferible a heredar cuando la relación es tiene un, no es un. Mantén atributos públicos simples al inicio y evita diseñar jerarquías profundas antes de tener un problema que las justifique.

Las clases no deben mezclar lectura de archivos, interfaz y reglas de dominio sin necesidad. Una buena separación sería: una función lee CSV, objetos Lectura representan datos validados y otra función prepara un informe. Así cada parte puede probarse con facilidad.

## Proyecto 13 · Modelo de confort

Crea las clases Aula y Medicion. Aula tendrá código, capacidad y ocupacion, con métodos registrar_entrada y plazas_libres que respeten límites. Medicion tendrá un Aula, temperatura y humedad, con un método estado que devuelva revisar temperatura, revisar humedad o adecuada. Añade representaciones útiles y pruebas assert para estados iniciales, capacidad llena, límites ambientales y dos aulas independientes.

Documenta en Markdown por qué Medicion contiene un Aula en lugar de heredar de Aula, qué atributos pueden cambiar y qué regla protege cada método. Ejecuta el proyecto desde un kernel reiniciado.

```python
class Aula:
    def __init__(self, codigo: str, capacidad: int = 30, ocupacion: int = 0):
        self.codigo = codigo
        self.capacidad = capacidad
        self.ocupacion = ocupacion

    def plazas_libres(self) -> int:
        return self.capacidad - self.ocupacion

    def registrar_entrada(self) -> bool:
        if self.ocupacion < self.capacidad:
            self.ocupacion += 1
            return True
        return False

    def __repr__(self) -> str:
        return f"Aula(codigo='{self.codigo}' capacidad={self.capacidad} ocupacion={self.ocupacion})"

class Medicion:
    def __init__(self, aula: Aula, temperatura: float, humedad: float):
        self.aula = aula
        self.temperatura = temperatura
        self.humedad = humedad

    def estado(self) -> str:
        if self.temperatura < 18 or self.temperatura > 27:
            return "revisar temperatura"

        if self.humedad < 30 or self.humedad > 65:
            return "revisar humedad"

        return "adecuada"

    def __repr__(self) -> str:
        return f"Medicion(aula={self.aula.codigo} temperatura={self.temperatura} humedad={self.humedad})"



# Estados iniciales
aula1 = Aula("A-01")
assert aula1.ocupacion == 0
assert aula1.plazas_libres() == 30

# Capacidad llena
aula_llena = Aula("A-02", capacidad=2, ocupacion=2)
assert aula_llena.registrar_entrada() is False

assert Medicion(aula1, 18, 50).estado() == "adecuada"
assert Medicion(aula1, 27, 50).estado() == "adecuada"
assert Medicion(aula1, 22, 30).estado() == "adecuada"
assert Medicion(aula1, 22, 65).estado() == "adecuada"
assert Medicion(aula1, 17, 50).estado() == "revisar temperatura"
assert Medicion(aula1, 22, 70).estado() == "revisar humedad"

# Dos aulas independientes
aula2 = Aula("A-03")
aula3 = Aula("A-04")
aula2.registrar_entrada()
assert aula2.ocupacion == 1
assert aula3.ocupacion == 0

```

## Por qué Medicion contiene un Aula en lugar de heredar de Aula
- Medicion contiene un objeto Aula porque una medición no es un aula, solo debe pertenecer a un aula para conocer sus datos.

## Atributos que pueden cambiar
- En `Aula`: `ocupacion`.
- En `Medicion`: `temperatura` y `humedad`

# Regla de cada método:
- `plazas_libres()`: calcula las plazas disponibles.
- `registrar_entrada()`: comprueba que no se supere la capacidad del aula.
- `estado()`: comprueba que la temperatura esté entre 18 y 27 y la humedad entre 30 y 65.
