## 1. Operaciones vectorizadas y `ufuncs`

Una expresión como `temperaturas + 273.15` se aplica elemento a elemento. Funciones como `np.sqrt`, `np.log`, `np.clip` y `np.isfinite` trabajan sobre arrays completos. Antes de usarlas, revisa su dominio: por ejemplo, el logaritmo de un valor no positivo no es un resultado válido.

```python
temperaturas = np.array([20.5, 22.3, 21.1, 23.0])
kelvin = temperaturas + 273.15
centradas = temperaturas - temperaturas.mean()
print(kelvin)
print(centradas)
```

## 2. Broadcasting

Dos formas son compatibles si, al compararlas desde la derecha, coinciden o una de las dimensiones vale 1. Una matriz `(2, 3)` puede sumar un vector `(3,)`: el vector se aplica a cada fila. No cambies formas al azar para eliminar un error; primero identifica qué representa cada eje.

```python
medidas = np.array([[10, 20, 30], [12, 22, 32]], dtype=float)
correccion = np.array([0.5, -1.0, 0.0])
corregidas = medidas + correccion
print(corregidas)
```

## 3. Agregar por ejes

Sin `axis`, una agregación resume todo el array. En una matriz con filas como días y columnas como sensores, `axis=0` da un valor por sensor y `axis=1` da un valor por día. Elige el eje a partir de su significado, no de un resultado que parezca razonable.

```python
lecturas = np.array([[12, 15, 14], [13, 16, 15], [11, 14, 13]])
print("Media por sensor:", lecturas.mean(axis=0))
print("Media por día:", lecturas.mean(axis=1))
```

## 4. Condiciones y ausencias numéricas

`np.where` escoge valores según una condición. `np.isnan` detecta ausencias numéricas. Una normalización min–max exige que máximo y mínimo sean distintos. Ante un valor ausente debes decidir si excluir, imputar o detener el análisis; no conviene convertirlo en cero sin justificarlo.

```python
probabilidades = np.array([0.2, 0.65, 0.9, 0.4, 0.78])
clases = np.where(probabilidades >= 0.5, 1, 0)
print(clases)
print("Positivos:", clases.sum())
```
