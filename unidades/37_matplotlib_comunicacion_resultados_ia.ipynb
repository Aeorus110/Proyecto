```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 1. Curvas de valores ya calculados

Una **iteración** es un paso numerado de un proceso. Una **métrica** es una medida calculada siguiendo una regla definida. En este cuaderno, todos los valores ya están disponibles: una línea permite ver cómo cambian, pero no explica por qué cambian ni garantiza que el patrón se mantenga fuera de estos datos.

Cuando compares dos curvas, conserva la misma escala y nombra con claridad qué representa cada una. Si una curva se separa de otra, describe el momento y la magnitud observados, sin diagnosticar automáticamente un problema.

```python
iteraciones = np.arange(1, 9)
medida_referencia = np.array([0.42, 0.51, 0.58, 0.63, 0.67, 0.70, 0.72, 0.73])
medida_revision = np.array([0.39, 0.47, 0.53, 0.57, 0.60, 0.61, 0.60, 0.59])

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(iteraciones, medida_referencia, marker="o", label="Conjunto de referencia")
ax.plot(iteraciones, medida_revision, marker="s", linestyle="--", label="Conjunto de revisión")
ax.set(title="Medida precomputada por iteración", xlabel="Iteración", ylabel="Valor de la medida")
ax.set_xticks(iteraciones)
ax.set_ylim(0, 1)
ax.grid(axis="y", alpha=0.25)
ax.legend()
plt.show()

print("Diferencia final:", round(medida_referencia[-1] - medida_revision[-1], 2))
```

## 2. Matriz de confusión: contar pares de etiquetas

Una matriz de confusión resume cuántas veces una categoría real se asocia con una categoría predicha. Por convención en este curso, **las filas representan la categoría real y las columnas, la predicha**. La diagonal reúne los casos en que ambas coinciden; fuera de la diagonal aparecen desacuerdos.

Antes de interpretar una matriz, comprueba la orientación escrita en los ejes, el número total de casos y el reparto por fila. Una diagonal alta puede ocultar que una categoría poco frecuente tenga pocos casos. La matriz resume el conjunto disponible, no demuestra el comportamiento futuro del sistema.

```python
matriz = np.array([[24, 3, 1], [4, 18, 2], [1, 3, 20]])
clases = ["A", "B", "C"]

fig, ax = plt.subplots(figsize=(5.5, 4.5), layout="constrained")
imagen = ax.imshow(matriz, cmap="Blues", vmin=0, vmax=25)
fig.colorbar(imagen, ax=ax, label="Número de casos")
ax.set(title="Matriz de confusión precomputada", xlabel="Categoría predicha", ylabel="Categoría real")
ax.set_xticks(np.arange(3), clases)
ax.set_yticks(np.arange(3), clases)
for fila in range(matriz.shape[0]):
    for columna in range(matriz.shape[1]):
        color = "white" if matriz[fila, columna] > 12 else "black"
        ax.text(columna, fila, matriz[fila, columna], ha="center", va="center", color=color)
plt.show()

print("Casos evaluados:", matriz.sum())
print("Coincidencias de la diagonal:", np.trace(matriz))
```

## 3. Métricas por clase y barras agrupadas

Una métrica por clase permite detectar diferencias que un único promedio podría ocultar. En esta unidad empleamos nombres generales:

- **precisión**: de los casos asignados a una clase, qué parte coincide con esa clase;
- **cobertura**: de los casos reales de una clase, qué parte se localiza.

Ambas son proporciones entre 0 y 1. Mostrar la escala completa evita exagerar diferencias pequeñas. Las métricas deben ir acompañadas del número de casos y del método con que se calcularon, información que aquí se simplifica para practicar la visualización.

```python
clases = np.array(["A", "B", "C"])
precision = np.array([0.83, 0.75, 0.87])
cobertura = np.array([0.86, 0.72, 0.83])
x = np.arange(len(clases))
ancho = 0.36

fig, ax = plt.subplots(figsize=(7, 4))
barras_precision = ax.bar(x - ancho / 2, precision, ancho, label="Precisión")
barras_cobertura = ax.bar(x + ancho / 2, cobertura, ancho, label="Cobertura")
ax.set(title="Métricas precomputadas por clase", xlabel="Clase", ylabel="Proporción")
ax.set_xticks(x, clases)
ax.set_ylim(0, 1)
ax.bar_label(barras_precision, fmt="%.2f", padding=2)
ax.bar_label(barras_cobertura, fmt="%.2f", padding=2)
ax.grid(axis="y", alpha=0.25)
ax.legend()
plt.show()
```

## 4. Comunicar resultado, límite y siguiente paso

Una conclusión técnica útil tiene tres componentes:

1. **Resultado:** qué muestran exactamente la curva, matriz o barras.
2. **Límite:** qué no permiten asegurar los datos disponibles (muestra, categorías, periodo, método).
3. **Siguiente paso:** qué comprobación o dato adicional ayudaría a decidir.

Ejemplo: «La cobertura de B es la menor de estas tres clases. Los datos no indican si se debe a pocas observaciones, a una definición ambigua o a otro factor. Conviene revisar el número de casos y ejemplos de la fila B antes de proponer cambios».
