```python
import numpy as np

matriz = np.arange(1, 17).reshape(4, 4)
print(matriz)
```

## 1. Índices y cortes

NumPy usa índices desde cero y cortes cuyo límite final no se incluye, igual que las listas. En dos dimensiones se escribe `array[fila, columna]`. Una fila aislada suele devolver un array 1D; un corte de una sola fila, como `1:2`, mantiene las dos dimensiones. Comprueba `shape` cuando esa diferencia importe.

```python
print("Elemento de fila 1, columna 2:", matriz[1, 2])
print("Segunda fila:", matriz[1])
print("Primera columna:", matriz[:, 0])
print("Bloque central:")
print(matriz[1:3, 1:3])
```

## 2. Máscaras booleanas

Una comparación aplicada a un array produce una máscara de valores `True` y `False` de igual forma. La máscara permite seleccionar y asignar todos los elementos que cumplen una regla. Es una alternativa clara a un bucle cuando la regla es igual para cada valor.

```python
pm25 = np.array([12.7, 18.4, 23.5, 15.9, 28.1])
mascara_alta = pm25 > 20
print(mascara_alta)
print("Valores altos:", pm25[mascara_alta])
```

## 3. Modificación vectorizada

Una máscara también permite asignar nuevos valores. Modifica una copia cuando quieras preservar los datos de entrada. Antes de escribir una condición, decide qué debe ocurrir en los límites: por ejemplo, `< 5` no incluye exactamente el valor 5.

```python
precios = np.array([5.5, 12.0, 7.8, 3.5, 15.0, 2.3])
precios_ajustados = precios.copy()
precios_ajustados[precios_ajustados < 5] = 5
print(precios)
print(precios_ajustados)
```

## 4. Vistas y copias

Un corte suele crear una **vista**, que comparte datos con el array original. Cambiarla puede cambiar el origen. `.copy()` crea datos independientes. No dependas de recordar todos los casos de indexación: cuando una transformación deba ser independiente, usa una copia y comprueba el original con una aserción o comparación.

```python
original = np.arange(9).reshape(3, 3)
vista = original[:, :2]
copia = original[:, :2].copy()
vista[0, 0] = -1
copia[0, 1] = -2
print("Original:")
print(original)
print("Copia:")
print(copia)
```

## Práctica guiada

Con `np.arange(1, 17).reshape(4, 4)`, extrae segunda fila, primera columna y bloque central. Después multiplica por 10 sólo los valores mayores que 10. Explica qué conceptos de Python reaparecen y por qué no necesitas un bucle por cada celda.

```python
x = np.arange(1, 17).reshape(4, 4)
segunda_fila = x[1]
primera_columna = x[:, 0]
bloque_central = x[1:3, 1:3]
x[x > 10] *= 10
print(segunda_fila)
print(primera_columna)
print(bloque_central)
print(x)
```
