```python
import numpy as np

print("NumPy:", np.__version__)
```

## 1. Del listado al array

Una lista puede contener objetos de distinto tipo. Un `ndarray` tiene valores de un tipo común y una forma definida. No sustituye a las listas en todos los casos, pero resulta apropiado para bloques de mediciones numéricas. Las filas de una matriz deben tener igual longitud.

```python
temperaturas_lista = [12.4, 13.1, 14.8]
temperaturas = np.array(temperaturas_lista, dtype=float)
print(temperaturas)
print("shape:", temperaturas.shape)
print("ndim:", temperaturas.ndim)
print("size:", temperaturas.size)
print("dtype:", temperaturas.dtype)
```

## 2. Formas, ejes y constructores

En una matriz `(filas, columnas)`, el eje 0 recorre filas y el eje 1 columnas. `zeros`, `ones`, `full`, `arange`, `linspace` e `eye` expresan cómo se construyen los datos. `empty` reserva memoria, pero sus valores iniciales no son fiables: asígnalos antes de leerlos.

```python
lecturas = np.zeros((3, 4), dtype=float)
horas = np.arange(0, 24, 6)
referencias = np.linspace(10, 20, num=5)
identidad = np.eye(3, dtype=int)
print(horas)
print(referencias)
print(identidad)
```

## 3. Tipos y conversiones prudentes

El tipo determina precisión, rango y memoria. Dividir enteros suele producir decimales; convertir después a entero descarta la parte decimal. `astype` debe reflejar una decisión explícita, nunca ocultar información.

```python
conteos = np.array([12, 15, 9], dtype=np.int64)
promedios = conteos / 2
print(conteos.dtype, conteos.itemsize)
print(promedios, promedios.dtype)
```

```python
# 25.01

array = np.array([3.14, 2.71, 1.41, 0.0, -1.0])
print("Shape:", array.shape)  # (5,) -> Vector 1D de 5 elementos
print("Ndim:", array.ndim)    # 1 -> Una sola dimensión
print("Size:", array.size)    # 5 -> Total de elementos
print("Dtype:", array.dtype)  # float64 -> NumPy infiere float al haber decimales
```

```python
# 25.02. Construye una matriz 5×5 de centenas y cambia sólo la diagonal principal a cero. Recupera indexación básica de listas.

# 1. Creamos una matriz 5x5 de ceros con np.zeros y le sumamos 100
matriz = np.zeros((5, 5), dtype=int) + 100

# 2. Ponemos la diagonal a cero usando el bucle e indexación básica
for i in range(5):
    matriz[i, i] = 0
print(matriz)
```

```python
# 25.03. Genera los múltiplos de 3 de 0 a 33 y organízalos en una matriz 3×4. Comprueba tamaño y forma.
multiplos = np.arange(0, 34, 3).reshape(3, 4)
print(multiplos)
print("Shape:", multiplos.shape)
print("Size:", multiplos.size)
```

```python
# 25.04.Representa 4 sensores, 6 horas y 2 variables en un array de ceros de forma `(4, 6, 2)`. Explica cada eje.
array = np.zeros((4, 6, 2), dtype=float)
print("Forma (shape)", array.shape)
print("Dimensiones (ndim)", array.ndim)
print("Tamaño (size)", array.size)
```

```python
# 25.05. Compara dividir un array entero entre 2 antes y después de `astype(int)`. Explica la pérdida de información.

conteos = np.array([12, 15, 9], dtype=np.int64)

division_directa = conteos / 2
print("División directa:", division_directa, "| Tipo:", division_directa.dtype)

despues_de_astype = (conteos / 2).astype(int)
print("Después de astype(int):", despues_de_astype, "| Tipo:", despues_de_astype.dtype)

antes_astype = conteos.astype(int) / 2
print("Antes de astype(int):", antes_astype, "| Tipo:", antes_astype.dtype)
```
