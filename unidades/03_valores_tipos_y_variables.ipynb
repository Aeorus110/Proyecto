## 1. Valores, objetos y tipos

Un **valor** es información concreta, como `42`, `3.5`, `"Murcia"` o `True`. En Python los datos se representan mediante **objetos**. Para empezar, puedes pensar que un objeto es el valor con el que trabaja el intérprete y que conoce qué operaciones puede realizar. Más adelante veremos objetos más complejos, pero la misma idea se mantiene.

Cada objeto tiene un **tipo**. El tipo clasifica el dato y define operaciones apropiadas. Por ejemplo, sumar dos enteros tiene sentido; sumar directamente un entero y un texto no, porque Python no puede decidir si deseabas realizar una suma numérica o construir un mensaje. Python evita esa ambigüedad generando un error en vez de adivinar.

La función `type(valor)` muestra el tipo. No hay que memorizar la representación interna de cada tipo: la utilidad de inspeccionarlo es comprobar nuestras suposiciones cuando un resultado sorprende.

````python
print(type(42))
print(type(3.5))
print(type("Murcia"))
print(type(True))
print(type(None))
````

* type(42) -> int (Número entero)
* type(3.5) -> float (Número con decimales)
* type(Murcia) -> str (Cadena de texto)
* type(True) -> bool (Valor booleano)
* type(None) -> NoneType (Ausencia de valor)

### Tipos básicos de esta unidad

| Tipo | Ejemplos | Para qué se usa |
| --- | --- | --- |
| `int` | `0`, `-12`, `250` | Cantidades enteras, índices, recuentos. |
| `float` | `3.14`, `-0.5`, `2.0` | Medidas y cálculos con parte decimal. |
| `str` | `"hola"`, `'A-12'` | Texto, códigos y datos leídos como caracteres. |
| `bool` | `True`, `False` | Respuestas sí/no y condiciones. |
| `NoneType` | `None` | Ausencia intencionada de un valor. |

`True`, `False` y `None` empiezan por mayúscula: son palabras reservadas de Python, no nombres que podamos cambiar. El texto debe ir entre comillas. `"18"` es texto, mientras que `18` es un número entero; que ambos muestren los mismos caracteres no los convierte en el mismo tipo de dato.

## 2. Expresiones y resultados

Una **expresión** es una pieza de código que Python evalúa para obtener un valor. Un literal como `8`, un nombre como `temperatura`, una operación como `8 + 2` y una llamada como `type(8)` son expresiones. Python evalúa primero las partes necesarias y devuelve el resultado.

Los operadores numéricos más habituales son `+` (suma), `-` (resta), `*` (multiplicación), `/` (división real), `//` (división entera), `%` (resto) y `**` (potencia). Los paréntesis indican qué debe calcularse antes. Sin paréntesis, Python aplica reglas de prioridad similares a las matemáticas: potencia, multiplicación/división y suma/resta. Cuando una expresión es importante, los paréntesis mejoran la legibilidad incluso si no fueran estrictamente necesarios.

````python
print(9 + 3)
print(9 / 4)       # / devuelve un float , 2.25
print(9 // 4)      # // descarta la parte decimal , 2
print(9 % 4)       # % devuelve el resto , 2
print(2 + 3 * 4)
print((2 + 3) * 4)
````

* 7 + 3 -> 10 (Suma)
* 7 / 2 -> 3.5 (División real)
* 7 // 2 -> 3 (División entera que descarta la parte decimal)
* 7 % 2 -> 1 (Resto o módulo de la división)
* 2 + 3 * 4 -> 14 (Precedencia de operadores: multiplica antes de sumar)
* (2 + 3) * 4 -> 20 (Los paréntesis priorizan la operación interna)

### Una nota sobre números reales

Los `float` se almacenan en formato binario y no pueden representar exactamente muchos números decimales. Por ello, una operación aparentemente sencilla puede mostrar un resultado como `0.30000000000000004`. No es un fallo de Python: es una consecuencia del formato numérico usado por la mayoría de ordenadores. Para medidas, cálculos científicos y visualizaciones suele ser aceptable; para importes monetarios se usan normalmente estrategias específicas que estudiaremos cuando sean necesarias. Por ahora, no compares resultados decimales extensos por su representación visual sin pensar en el contexto.

## 3. Variables: nombres que se refieren a valores

Una **variable** es un nombre que se asocia a un objeto. La instrucción de asignación tiene la forma `nombre = expresión`: Python evalúa primero la expresión de la derecha y después asocia el resultado al nombre de la izquierda. Por esta razón, `total = precio + impuesto` se lee como «calcula la suma y guárdala con el nombre `total`», no como una igualdad matemática.

En Python no se declara por adelantado el tipo de una variable. El nombre puede referirse en distintos momentos a objetos de tipos diferentes, aunque hacerlo sin necesidad suele dificultar la lectura. Una buena práctica es que un nombre conserve un significado claro durante el programa.

````python
temperatura_aula = 23.5
limite_superior = 27
margen = limite_superior - temperatura_aula

print("Temperatura:", temperatura_aula)
print("Margen hasta el límite:", margen)
````

* Temperatura -> 23.5
* Margen es el resultado de 27 - 23.5 = 3.5

### Reglas y criterios para nombrar

Un nombre puede contener letras, dígitos y guiones bajos, pero no puede empezar por un dígito. Distingue mayúsculas y minúsculas: `temperatura` y `Temperatura` son nombres distintos. Tampoco puede coincidir con una palabra reservada, como `if` o `class`.

Usa `snake_case`: minúsculas y guiones bajos, por ejemplo `numero_alumnos` o `precio_final`. Evita nombres sin significado como `x`, `dato` o `cosa` cuando un nombre más descriptivo aclare el propósito. También evita sobrescribir nombres de funciones incorporadas como `list`, `str` o `type`; si lo haces, podrías dejar de poder usarlas en el resto del cuaderno.

### ¿Copiar o referirse al mismo objeto?

En `copia = original`, el nombre `copia` pasa a referirse inicialmente al mismo objeto que `original`; la asignación no implica necesariamente una copia independiente. Con los números y el texto, que trataremos aquí, esto suele ser sencillo porque no se modifican en el lugar. Al estudiar listas y diccionarios veremos por qué esta idea es importante con objetos mutables.

`id(valor)` ofrece un identificador de implementación que puede ayudarte a observar referencias durante una prueba. No es un dato que debas guardar ni comparar entre ejecuciones. Su objetivo didáctico es mostrar que dos nombres pueden apuntar al mismo objeto.

````python
aula = "A-12"
aula_asignada = aula

print(aula_asignada)
print(aula is aula_asignada)
print(id(aula) == id(aula_asignada))
````

Aula asignada hace referencia a aula, por lo que el resultado de aula y aula_asignada van a ser iguales

## 4. Conversión explícita de tipos

Muchos datos llegan como texto: una lectura desde un archivo, una respuesta escrita por una persona o un valor recibido de una página web. Antes de calcular con ellos, hay que convertirlos si realmente representan números. Las funciones `int()`, `float()`, `str()` y `bool()` realizan conversiones explícitas.

La conversión debe tener sentido y puede fallar. `int("24")` funciona; `int("24.5")` no, porque ese texto no representa un entero. En ese caso puede usarse `float("24.5")`. `float("veinticuatro")` tampoco funciona: antes habría que corregir o rechazar el dato. Python no interpreta palabras numéricas por nosotros.

````python
lectura_texto = "18"
lectura_numerica = float(lectura_texto)

print(lectura_numerica + 1)
print(str(lectura_numerica) + " °C")
print(type(lectura_numerica))
````

* 24.5 Era de tipo str, pero fue convertido a float por lo que si puede sumarse con 1
* El resultado de la suma es convertido de forma temporal a str para poder sumar con el caracter "°C"
* Muestra el tipo de la variable que era float, ya que el print no modifica el valor original

### Booleanos y ausencia de valor

Un booleano solo puede ser `True` o `False`. Será la base de las decisiones de la Unidad 04. No confundas `True` con el texto `"True"`: el primero es un booleano y el segundo una cadena de caracteres.

`None` significa que no hay un valor intencionadamente. No equivale a cero, a `False` ni a un texto vacío. Por ejemplo, una temperatura de `0` °C es una medida válida; una lectura `None` puede significar que el sensor no respondió. Esta distinción evita convertir un dato ausente en un resultado engañoso.

````python
sensor_activo = True
ultima_lectura = None

print(sensor_activo, type(sensor_activo))
print(ultima_lectura, type(ultima_lectura))
````

* sensor_activo es un booleano con valor True
* ultima_lectura es None que no tiene valor definido

## Ejercicio 03.02 · Expresiones y asignación

Una sala tiene 28 puestos, de los que 19 están ocupados. Escribe código que guarde ambos números con nombres claros y calcule:

1. puestos libres;
2. porcentaje de ocupación como número decimal;
3. porcentaje de ocupación como texto seguido de `%`.

Muestra cada resultado y usa paréntesis cuando ayuden a comprender la fórmula.

````python
sala = 28
visitantes = 19
puestos_libres = sala - visitantes
ocupacion_decimal = f"{(visitantes*100)/28:.2f}"
ocupacion_texto = f"{(visitantes*100)/28:.2f}%"

print("Puestos Libres: " + str(puestos_libres))
print("Ocupacion en forma decimal: " + ocupacion_decimal)
print("Ocupacion en forma de texto: " + ocupacion_texto)
````
