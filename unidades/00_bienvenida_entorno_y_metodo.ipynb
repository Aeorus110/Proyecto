## 2. Entorno virtual y recursos

El entorno virtual `.venv` aísla las herramientas de este curso. Debe activarse antes de abrir JupyterLab. `Path.cwd()` representa la carpeta actual; `exists()` indica si un recurso existe.

```python
from pathlib import Path
import platform
import sys

print("Python:", sys.version.split()[0])
print("Sistema:", platform.system())
raiz = Path.cwd()
for recurso in ("unidades", "datos", "modulos", "requirements.txt"):
    print(f"{recurso:16} -> {(raiz / recurso).exists()}")
```
