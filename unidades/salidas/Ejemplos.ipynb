## 1. Python: recorrer, contar y filtrar

### Reto

Dada una lista de temperaturas, crea una lista con las que son mayores o iguales a 20 y cuenta cuantas cumplen la condicion.

### Solucion

````python
temperaturas = [18, 21, 20, 16, 25]
adecuadas = []

for temperatura in temperaturas:
    if temperatura >= 20:
        adecuadas.append(temperatura)

print(adecuadas)
print(len(adecuadas))
````

**Resultado esperado:** `[21, 20, 25]` y `3`.

**Caso limite:** si `temperaturas` esta vacia, el resultado es una lista vacia y el contador es cero.

## 2. Python: contar con un diccionario

### Reto

Cuenta cuantas veces aparece cada etiqueta.

### Solucion

````python
etiquetas = ["web", "movil", "web", "web", "movil"]
conteos = {}

for etiqueta in etiquetas:
    conteos[etiqueta] = conteos.get(etiqueta, 0) + 1

print(conteos)
````

**Resultado esperado:** `{'web': 3, 'movil': 2}`.

**Caso limite:** con una lista vacia, `conteos` queda vacio.

## 3. Python: funcion, retorno y validacion

### Reto

Escribe una funcion que calcule la media. Si no recibe valores, debe devolver `None`.

### Solucion

````python
def calcular_media(valores):
    if len(valores) == 0:
        return None
    return sum(valores) / len(valores)

datos = [10, 12, 14]
assert calcular_media(datos) == 12
assert calcular_media([]) is None
print(calcular_media(datos))
````

**Resultado esperado:** `12.0`.

**Recuerda:** `return` devuelve un valor; `print` solo lo muestra.

## 6. Pandas: agrupar y resumir

### Reto

Calcula la temperatura media por zona.

### Solucion

````python
import pandas as pd

datos = pd.DataFrame({
    "zona": ["norte", "sur", "norte", "sur"],
    "temperatura": [18, 20, 22, 24],
})
resumen = datos.groupby("zona")["temperatura"].mean()
print(resumen)
````

**Resultado esperado:** `norte` tiene media `20.0` y `sur`, `22.0`.

**Comprueba:** la media por grupo no es lo mismo que la media global.

````python
class Aula:
    def __init__(self, capacidad, ocupacion=0):
        self.capacidad = capacidad
        self.ocupacion = ocupacion

    def plazas_libres(self):
        return self.capacidad - self.ocupacion

aula = Aula(30, 12)
assert aula.plazas_libres() == 18
````

### 31. NumPy: normalizacion segura

**Reto:** Normaliza valores entre 0 y 1 y trata el caso en que todos son iguales.

**Idea clave:** Un rango cero haria indefinida la division; se elige explicitamente devolver ceros.


````python
import numpy as np

def normalizar(valores):
    valores = np.asarray(valores, dtype=float)
    rango = valores.max() - valores.min()
    if rango == 0:
        return np.zeros_like(valores)
    return (valores - valores.min()) / rango

assert np.allclose(normalizar([5, 10, 15]), [0, 0.5, 1])
assert np.allclose(normalizar([4, 4]), [0, 0])
````

### 41. Seaborn: correlacion y mascara triangular

**Reto:** Calcula correlaciones y oculta la mitad superior redundante.

**Idea clave:** La correlacion describe asociacion lineal; no demuestra causalidad.


````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

datos = pd.DataFrame({"a": [1, 2, 3], "b": [2, 4, 6], "c": [3, 2, 1]})
corr = datos.corr()
mascara = np.triu(np.ones_like(corr, dtype=bool), k=1)
sns.heatmap(corr, mask=mascara, vmin=-1, vmax=1, center=0, annot=True, cmap="vlag")
plt.show()
````
