```python
import numpy as np
```

## 1. Cambiar forma sin cambiar datos

`reshape` sólo funciona si el número total de elementos coincide. `-1` permite que NumPy calcule una dimensión, pero sólo debe usarse si puedes explicar qué tamaño resultará. `ravel` intenta devolver una vista aplanada; `flatten` crea una copia. Esa diferencia importa si modificarás el resultado.

```python
valores = np.arange(12)
matriz = valores.reshape(3, 4)
print(matriz)
print(matriz.reshape(2, -1))
print(matriz.ravel().shape)
```

## 2. Transponer, concatenar y dividir

`.T` intercambia ejes en una matriz 2D. `np.vstack` agrega filas; `np.hstack` agrega columnas; `np.concatenate` permite indicar el eje. Las formas compatibles no garantizan unidades compatibles: antes de combinar, verifica que cada columna representa la misma variable y escala.

```python
manana = np.array([[12, 60], [14, 55]])
tarde = np.array([[20, 45], [22, 40]])
dia = np.vstack([manana, tarde])
print(dia)
print(dia.T)
```

## 3. Preparación numérica

En muchas tareas posteriores, las filas son observaciones y las columnas son variables. Si la última columna es la variable que quieres estudiar, puedes separar `X` e `y`. Esta separación no crea un modelo ni convierte una asociación en una predicción: sólo organiza los datos para el siguiente paso.

```python
datos = np.array([[12.0, 60.0, 18.0], [14.0, 55.0, 15.0], [20.0, 45.0, 22.0]])
x = datos[:, :2]
y = datos[:, 2]
print(x.shape, y.shape)
```

## Práctica guiada

Crea 24 valores, organízalos en 6 filas y 4 columnas, divide la matriz en dos bloques de tres filas y vuelve a unirlos. Comprueba que el orden y la forma se conservan. Después explica qué significaría cada fila y columna si fueran lecturas de sensores.

```python
matriz = np.arange(24).reshape(6, 4)
superior, inferior = np.vsplit(matriz, 2)
reconstruida = np.vstack([superior, inferior])
assert np.array_equal(matriz, reconstruida)
print(superior)
```
