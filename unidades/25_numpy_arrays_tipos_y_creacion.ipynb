## 1. Del listado al array

Una lista puede contener objetos de distinto tipo. Un `ndarray` tiene valores de un tipo común y una forma definida. No sustituye a las listas en todos los casos, pero resulta apropiado para bloques de mediciones numéricas. Las filas de una matriz deben tener igual longitud.

```python
temperaturas_lista = [12.4, 13.1, 14.8]
temperaturas = np.array(temperaturas_lista, dtype=float)
print(temperaturas)
print("shape:", temperaturas.shape)
print("ndim:", temperaturas.ndim)
print("size:", temperaturas.size)
print("dtype:", temperaturas.dtype)
```

## 2. Formas, ejes y constructores

En una matriz `(filas, columnas)`, el eje 0 recorre filas y el eje 1 columnas. `zeros`, `ones`, `full`, `arange`, `linspace` e `eye` expresan cómo se construyen los datos. `empty` reserva memoria, pero sus valores iniciales no son fiables: asígnalos antes de leerlos.

```python
lecturas = np.zeros((3, 4), dtype=float)
horas = np.arange(0, 24, 6)
referencias = np.linspace(10, 20, num=5)
identidad = np.eye(3, dtype=int)
print(horas)
print(referencias)
print(identidad)
```

## 3. Tipos y conversiones prudentes

El tipo determina precisión, rango y memoria. Dividir enteros suele producir decimales; convertir después a entero descarta la parte decimal. `astype` debe reflejar una decisión explícita, nunca ocultar información.

```python
conteos = np.array([12, 15, 9], dtype=np.int64)
promedios = conteos / 2
print(conteos.dtype, conteos.itemsize)
print(promedios, promedios.dtype)
```
