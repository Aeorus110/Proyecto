## 1. Condiciones y booleanos

Una **condición** es una expresión cuyo resultado es `True` o `False`. En la unidad anterior vimos que esos dos valores forman el tipo `bool`. Ahora los utilizaremos para decidir qué instrucciones se ejecutan.

Las comparaciones más frecuentes son `==` (igual), `!=` (distinto), `<` (menor), `<=` (menor o igual), `>` (mayor) y `>=` (mayor o igual). Es muy importante distinguir `=` de `==`: el primero asigna un valor a un nombre; el segundo compara dos valores. En una condición se usa `==`, nunca `=`.

````python
from hmac import trans_36
from string import punctuation

temperatura = 26.5
limite = 27

print(temperatura < limite)
print(temperatura == limite)
print(temperatura >= limite)
print(type(temperatura < limite))
````

### Comparar el dato correcto

Una condición solo tiene sentido si los valores comparados representan la misma clase de información. Comparar una temperatura numérica con un límite numérico es razonable. Comparar el texto `"27"` con el número `27` no responde a esa pregunta: primero habría que convertir el texto a número, como se estudió en la Unidad 03.

Para comprobar si un dato ausente está representado por `None`, se emplea normalmente `is None`, no `== None`. `is` pregunta si se trata del mismo objeto especial `None`; en cambio, `==` compara igualdad de valor. Usaremos `is None` cuando sea necesario distinguir una ausencia de dato.

````python
lectura = None
print(lectura is None)
print(lectura is not None)
````

## 2. Combinar condiciones

A menudo una decisión necesita más de una comparación. Python proporciona operadores lógicos:

| Operador | Resultado | Ejemplo |
| --- | --- | --- |
| `and` | `True` solo si ambas condiciones son verdaderas. | `temperatura >= 18 and temperatura <= 27` |
| `or` | `True` si al menos una condición es verdadera. | `temperatura < 18 or temperatura > 27` |
| `not` | Invierte un booleano. | `not sensor_activo` |

Los paréntesis ayudan a mostrar la intención cuando una condición es larga. Es preferible una condición clara a una forma breve pero difícil de revisar. También es posible escribir `18 <= temperatura <= 27`; Python entiende esta expresión encadenada como una comprobación de intervalo.

````python
temperatura = 26.5
sensor_activo = True

en_rango = 18 <= temperatura <= 27
fuera_de_rango = temperatura < 18 or temperatura > 27
requiere_revision = (not sensor_activo) or fuera_de_rango

print(en_rango)
print(fuera_de_rango)
print(requiere_revision)
````

## 3. Elegir un camino con `if`

La instrucción `if` ejecuta un bloque de código solamente cuando su condición es verdadera. La forma básica es:

```python
if condicion:
    instruccion_si_verdadera
```

Los dos puntos al final de la condición y la sangría de las líneas siguientes son obligatorios. Python usa la sangría para saber qué instrucciones pertenecen al bloque. Por convención se utilizan cuatro espacios. Cuando la condición es `False`, el bloque sangrado se omite y la ejecución continúa después de él.

````python
temperatura = 28.0

if temperatura > 27:
    print("Temperatura alta: revisar ventilación.")

print("Lectura procesada.")
````

### Alternativas con else y elif

else define qué hacer cuando la condición de un if es falsa. elif permite comprobar otra condición si las anteriores no se cumplieron. En una cadena de decisiones, Python examina las condiciones de arriba abajo y ejecuta solamente el primer bloque cuya condición sea verdadera. Después salta al final de la cadena.

El orden importa. Si una condición amplia aparece antes que otra más específica, puede impedir que la segunda se alcance. Por ejemplo, si preguntas primero si la temperatura es mayor que 18 y después si es mayor que 27, nunca llegarás al segundo caso para temperaturas altas. Ordena de los casos específicos a los generales o diseña intervalos que no se solapen.

````python
temperatura = 17.5

if temperatura < 18:
    estado = "baja"
elif temperatura > 27:
    estado = "alta"
else:
    estado = "adecuada"

print("Temperatura", estado)
````

### Diseñar intervalos sin huecos

Antes de escribir código, anota qué debe ocurrir en cada intervalo. Para clasificar una temperatura como baja si es menor que 18, adecuada entre 18 y 27 ambos incluidos, y alta si es mayor que 27, los valores frontera son precisamente 18 y 27. En el ejemplo anterior, el bloque else cubre el intervalo restante porque las dos condiciones anteriores ya han descartado los extremos.

No hay una regla universal sobre incluir o excluir un límite: depende del enunciado. Lo importante es expresarlo de forma explícita, probarlo y no dejar huecos ni solapamientos involuntarios.

## 4. Depuración inicial

**Depurar** es investigar por qué un programa no se comporta como se espera y corregir la causa. No consiste en cambiar muchas líneas hasta que desaparezca un síntoma. Un método breve y repetible es:

1. **Reproducir:** ejecutar el programa con un dato concreto que muestre el problema.
2. **Describir lo esperado y lo obtenido:** por ejemplo, «con 18 esperaba adecuada y se muestra baja».
3. **Reducir:** aislar las pocas líneas y datos que intervienen.
4. **Observar:** usar print temporal para mostrar valores y resultados de condiciones.
5. **Formular una hipótesis:** decidir qué regla o dato puede explicar el resultado.
6. **Cambiar lo mínimo y volver a probar:** comprobar también los casos cercanos.

Las trazas de ejecución explicadas en la Unidad 02 sirven para errores que detienen el programa. En decisiones, muchos errores son de lógica: el código termina, pero recorre un camino incorrecto. Por eso las pruebas con datos elegidos conscientemente son indispensables.

````python
temperatura = 18
es_baja = temperatura < 18
es_alta = temperatura > 27

print("temperatura:", temperatura)
print("es_baja:", es_baja)
print("es_alta:", es_alta)
````

### Casos de prueba

Para una regla con límites, no basta una prueba que parece normal. Diseña al menos un caso normal dentro de cada categoría; casos límite, como 18 y 27; valores justo fuera de ellos, como 17.9 y 27.1; y un dato inválido o ausente cuando el problema pueda recibirlo.

Esta lista no sustituye a pensar en el problema, pero evita muchos errores típicos. En una aplicación real también se validaría que el dato recibido tiene el tipo y la unidad esperados.

````python
t1 = 18
t2 = 27
t3 = 17.9
t4 = 27.1
dato_invalido = -20

if t1 >= 18 and t2 <= 27:
    print("Temperatura dentro del rango")
elif t3 < 18 and t4 > 27:
    print("Temperatura fuera del rango")

if dato_invalido < 19:
    print("Dato inválido")
````

- Si está activo, la temperatura es adecuada y la humedad es mayor que 65, el estado será revisar humedad.
