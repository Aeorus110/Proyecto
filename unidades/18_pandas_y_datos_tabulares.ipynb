## 1. Por qué usar una tabla para datos

Una tabla organiza observaciones en filas y variables en columnas. Una observación es un caso concreto: por ejemplo, una lectura de una estación ambiental en una fecha y una hora. Una variable es una propiedad medida o registrada de ese caso, como la temperatura, la zona o la concentración de partículas.

Esta estructura no hace que los datos sean correctos por sí sola. Antes de sacar conclusiones hay que conocer su procedencia, qué unidad usa cada variable, qué población representa y qué limitaciones tiene. En esta unidad inspeccionaremos; limpiar, filtrar y analizar llegará después.

## 3. Series: una columna con etiqueta

Una Series es una secuencia de valores con un índice. Se parece a una lista, pero cada valor queda asociado a una etiqueta de posición. Una columna de un DataFrame será normalmente una Series. Los valores de una Series pueden ser numéricos, texto, fechas u otros objetos, aunque es conveniente que una misma variable tenga un significado y tipo coherentes.

```python
temperaturas = pd.Series([10.8, 16.2, 12.1], name='temperatura_c')
temperaturas
```

```python
temperaturas.index
temperaturas.name
temperaturas.dtype
```

El índice inicial es 0, 1 y 2. No es una variable medida: es una etiqueta que Pandas usa para identificar posiciones. Más adelante veremos índices personalizados y selección; por ahora conviene no confundirlo con una columna llamada, por ejemplo, estacion_id.

## 4. DataFrame: una tabla de variables relacionadas

Un DataFrame reúne varias Series que comparten un índice. En un buen DataFrame cada fila representa el mismo tipo de observación y cada columna representa una variable. Una tabla de lecturas ambientales puede contener fecha, hora, estación y medidas; todas describen la misma lectura.

Crear una tabla pequeña a mano es útil para comprender su forma. En situaciones reales se suele leer desde archivos, bases de datos o servicios web.

```python
lecturas_pequenas = pd.DataFrame(
    {
        'estacion_id': ['EST-01', 'EST-02', 'EST-03'],
        'zona': ['Norte', 'Centro', 'Sur'],
        'temperatura_c': [10.8, 11.7, 12.5],
    }
)
lecturas_pequenas
```

### Forma, tamaño y dimensiones

shape devuelve una pareja: primero número de filas y después número de columnas. size devuelve el número total de celdas y ndim indica el número de dimensiones. Un DataFrame es bidimensional incluso si tiene una sola fila o columna. No interpretes shape como una medida de calidad: una tabla grande puede ser poco útil y una pequeña puede ser excelente para aprender o tomar una decisión concreta.

```python
print('shape:', lecturas_pequenas.shape)
print('size:', lecturas_pequenas.size)
print('ndim:', lecturas_pequenas.ndim)
print('columnas:', list(lecturas_pequenas.columns))
```

## 5. El formato CSV

CSV significa valores separados por comas. Es un formato de texto, no una hoja de cálculo: puedes abrirlo con un editor, pero sus reglas importan. La primera línea suele ser la cabecera; cada línea siguiente representa una fila. Algunos CSV usan punto y coma, una codificación distinta de UTF-8 o símbolos especiales para valores ausentes.

Pandas no puede adivinar siempre estas decisiones. Si el resultado tiene una sola columna cuando esperabas varias, revisa primero el separador. Si aparecen caracteres extraños, revisa la codificación. Si una fecha aparece como texto, no es necesariamente un error: Pandas no debe asumir cómo interpretar cada cadena.

### Rutas relativas y reproducibilidad

Una ruta relativa describe un archivo respecto al directorio desde el que se ejecuta el cuaderno. Es preferible a una ruta absoluta de tu ordenador porque permite que otra persona clone o descargue el repositorio y ejecute el mismo trabajo. pathlib ayuda a construir rutas que funcionan en Windows, macOS y Linux.

El siguiente código acepta ejecutar JupyterLab desde la carpeta unidades o desde la raíz del proyecto. Observa la ruta que se muestra antes de leer: cuando un archivo no aparece, inspeccionar la ruta es más útil que repetir la misma celda.

```python
from pathlib import Path

ruta_datos = Path('..') / 'datos' / 'pandas' / '18_lecturas_estaciones.csv'
if not ruta_datos.exists():
    ruta_datos = Path('datos') / 'pandas' / '18_lecturas_estaciones.csv'

print('Ruta usada:', ruta_datos.resolve())
print('¿Existe?', ruta_datos.exists())
```

```python
df = pd.read_csv(ruta_datos)
df.head()
```

El nombre df es una abreviatura frecuente de DataFrame. Es aceptable en un ejemplo corto, pero en un análisis con varias tablas es mejor elegir nombres como lecturas, estaciones o mantenimiento. Un nombre debe explicar qué contiene, no solo su tipo técnico.

## 6. Inspección inicial: mirar antes de modificar

La inspección inicial responde preguntas sencillas y decisivas: ¿el archivo se ha leído?, ¿cuántas observaciones contiene?, ¿qué columnas hay?, ¿qué tipos ha inferido Pandas?, ¿qué ejemplos de filas aparecen al principio y al final? No responde todavía si los datos son representativos o si existe una relación causal entre variables.

head y tail muestran, por defecto, cinco filas. sample permite obtener filas al azar; usar random_state hace que el ejemplo sea repetible.

```python
print('Primeras filas:')
display(df.head(3))

print('Últimas filas:')
display(df.tail(2))

print('Muestra reproducible:')
display(df.sample(3, random_state=42))
```

### Columnas, tipos e información general

columns contiene los nombres de las variables. dtypes indica el tipo inferido en cada columna. info resume filas, columnas, valores no nulos, tipos y memoria. La salida object suele representar texto, aunque también puede señalar una columna heterogénea que merece revisión. No conviertas tipos solo porque parezcan inesperados: primero identifica el significado de la variable y el formato de origen.

```python
print('Forma:', df.shape)
print('Tamaño:', df.size)
print('Dimensiones:', df.ndim)
print('Columnas:', list(df.columns))
print('\nTipos:')
print(df.dtypes)
print('\nInformación general:')
df.info()
```

En este CSV fecha y hora se leen inicialmente como texto. No es un defecto: el archivo no guarda un tipo fecha universal. En la unidad 21 aprenderás a convertir y validar fechas. En esta unidad basta con detectar y describir este hecho correctamente.

### Resumen numérico y variedad categórica

describe resume columnas numéricas con cantidad, media, desviación típica, mínimo, cuartiles y máximo. Es una descripción, no una explicación. La media de temperatura no demuestra que una zona cause otra temperatura. Para texto o categorías, nunique cuenta valores distintos y value_counts cuenta frecuencia de cada valor.

Observa siempre la unidad: temperatura_c está en grados Celsius, humedad_pct en porcentaje y pm25_ug_m3 en microgramos por metro cúbico. Comparar sus números sin atender a las unidades puede llevar a conclusiones sin sentido.

```python
df.describe()
```

```python
print('Valores distintos por columna:')
print(df.nunique())
print('\nLecturas por zona:')
print(df['zona'].value_counts())
```

## Errores frecuentes y cómo abordarlos

- FileNotFoundError: muestra la ruta, comprueba el directorio de trabajo y evita rutas personales absolutas.
- Una sola columna inesperada: comprueba si el archivo usa coma, punto y coma o tabulador como separador.
- Caracteres extraños: identifica la codificación real antes de probar valores al azar.
- Tipo object donde esperabas números o fechas: inspecciona ejemplos, unidades y símbolos; no fuerces una conversión sin justificarla.
- Interpretar head como el conjunto completo: las primeras filas solo son una muestra ordenada, no una garantía sobre todas las demás.
- Modificar df durante la inspección: conserva una tabla original hasta que la unidad de limpieza te permita documentar las transformaciones.
