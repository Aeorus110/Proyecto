## 1. Iterable e iterador

Un iterable es un objeto del que se pueden obtener elementos uno tras otro: listas, tuplas, cadenas, rangos, diccionarios y conjuntos son ejemplos. Un iterador es el objeto que mantiene el estado del recorrido y entrega el siguiente elemento cuando se solicita. La función iter crea un iterador a partir de un iterable y next solicita el elemento siguiente.

Un bucle for realiza internamente esa secuencia de forma automática. Cuando no quedan elementos, next lanza StopIteration y for termina el recorrido sin mostrar ese detalle. Conocer este modelo ayuda a comprender por qué algunos objetos solo se pueden recorrer una vez.

```python
aulas = ["A-01", "A-02", "B-01"]
iterador = iter(aulas)
print(next(iterador))
print(next(iterador))
print(next(iterador))
```

### Agotamiento y repetición

Una lista conserva sus elementos y puede recorrerse tantas veces como se necesite; cada for crea un iterador nuevo. Un iterador ya consumido no vuelve al principio. Si necesitas los datos otra vez, crea un iterador nuevo desde el iterable original o guarda los resultados en una colección.

No conviertas un flujo grande en lista solo por comodidad. Antes pregunta si se necesita todo el resultado simultáneamente o si basta procesar un elemento cada vez. Esta decisión afecta al uso de memoria y, sobre todo, a la claridad de la intención.

```python
numeros = iter(range(3))
print(list(numeros))
print(list(numeros))
print(list(range(3)))
```

## 2. Generadores y evaluación perezosa

Un generador produce valores bajo demanda. Una función generadora contiene yield: cada yield entrega un valor y pausa el estado; la siguiente solicitud continúa desde ese punto. No construye todos los valores a la vez. Esto se denomina evaluación perezosa.

Los generadores son útiles para secuencias grandes, lectura progresiva de datos o transformaciones que solo se consumen una vez. No son automáticamente mejores: si necesitas indexar, modificar o recorrer varias veces el resultado, una lista puede ser más adecuada.

```python
def lecturas_pares(limite: int):
    numero = 0
    while numero <= limite:
        yield numero
        numero = numero + 2

generador = lecturas_pares(6)
print(next(generador))
print(list(generador))
```

### Expresiones generadoras

Una expresión generadora tiene una sintaxis parecida a una comprensión de lista, pero usa paréntesis. No crea una lista inmediatamente. Puede entregarse directamente a funciones como sum, any o all, que consumen sus elementos. Si se muestra el generador con print, se verá su representación, no todos sus valores; conviértelo a lista solo para inspeccionar ejemplos pequeños.

Cuidado: una expresión generadora se agota. Si la pasas a sum y después intentas convertirla a lista, ya no tendrá elementos.

```python
temperaturas = [18.0, 20.5, 22.0]
cuadrados = (temperatura ** 2 for temperatura in temperaturas)
print(sum(cuadrados))
print(list(cuadrados))
```

## 3. Comprensiones

Una comprensión crea una colección a partir de un iterable. La comprensión de lista tiene una expresión de resultado, un for y opcionalmente una condición. Resume un patrón de recorrido, transformación y filtrado que antes escribiríamos con un bucle y append.

Utilízala cuando cabe en una línea legible y expresa una operación sencilla. Si necesita varias condiciones, efectos secundarios o lógica difícil de explicar, usa un bucle normal con nombres intermedios. La brevedad no justifica ocultar el razonamiento.

```python
lecturas = [17.5, 18.0, 21.3, 28.0]
lecturas_adecuadas = [lectura for lectura in lecturas if 18 <= lectura <= 27]
mensajes = [f"{lectura:.1f} °C" for lectura in lecturas_adecuadas]
print(lecturas_adecuadas)
print(mensajes)
```

### Conjuntos y diccionarios por comprensión

Una comprensión de conjunto usa llaves y elimina duplicados. Una comprensión de diccionario produce pares clave y valor. Son útiles para normalizar etiquetas únicas o asociar un identificador con una transformación. Igual que los conjuntos y diccionarios normales, no deben usarse si un orden de posición forma parte del significado.

Comprueba siempre que las claves producidas sean únicas o que la sobrescritura sea intencionada. Si dos elementos generan la misma clave, el último valor reemplaza al anterior.

```python
codigos = ["a_01", "A-01", "b_02"]
normalizados = {codigo.upper().replace("_", "-") for codigo in codigos}
longitudes = {codigo: len(codigo) for codigo in normalizados}
print(normalizados)
print(longitudes)
```

## 4. Elegir el recurso correcto

Usa una lista cuando necesites conservar, indexar o recorrer varias veces todos los resultados. Usa un generador cuando proceses elementos progresivamente y no necesites almacenarlos todos. Usa una comprensión cuando la transformación o filtrado sea simple. Usa un bucle explícito cuando priorices pasos intermedios, depuración o claridad.

Durante la depuración, transforma temporalmente una comprensión en un bucle para imprimir valores intermedios. Después decide cuál versión comunica mejor la solución final.

**Ejercicio 16.01.** Crea un generador que produzca múltiplos de tres hasta un límite. Consúmelo con un bucle y explica por qué no se puede recorrer dos veces.

**Ejercicio 16.02.** Dada una lista de temperaturas, crea mediante comprensión otra con las adecuadas y otra con sus mensajes formateados. Reescribe una de ellas como bucle.

**Ejercicio 16.03.** Crea un conjunto por comprensión que normalice etiquetas y un diccionario por comprensión que asocie cada etiqueta con su longitud. Explica qué sucede si dos entradas normalizan igual.

```python
# Ejercicio 16.01 Crea un generador que produzca múltiplos de tres hasta un límite. Consúmelo con un bucle y explica por qué no se puede recorrer dos veces.
def multiplos_de_tres(limite:int):
    numero = 3
    while numero <= limite:
        yield numero
        numero += 3

generador = multiplos_de_tres(10)
resultado = [numero for numero in generador]
assert resultado == [3,6,9]

segundo_recorrido = list(generador)
assert segundo_recorrido == []

# Un generador solo avanza hacia delante, por lo que una vez que se acaba el bucle se acaban los yield. Para volver a recoger los datos habría que llamar a la función de nuevo
```

```python
# Ejercicio 16.02 Dada una lista de temperaturas, crea mediante comprensión otra con las adecuadas y otra con sus mensajes formateados. Reescribe una de ellas como bucle.
temperaturas = [17.5, 18, 21.3, 28]
temperaturas_adecuadas_compresion = [temperatura for temperatura in temperaturas if 18 <= temperatura <= 27]
mensajes_formateados = [f"{temperatura:.1f} ºC" for temperatura in temperaturas_adecuadas_compresion]

# Pruebas
assert temperaturas_adecuadas_compresion == [18, 21.3]
assert mensajes_formateados == ["18.0 ºC", "21.3 ºC"] # Con 18 da error. Se debe poner el 18.0 para que no de error

temperaturas_adecuadas_bucle = []
for temperatura in temperaturas:
    if 18 <= temperatura <= 27:
        temperaturas_adecuadas_bucle.append(temperatura)

assert temperaturas_adecuadas_bucle == [18, 21.3]
```

```python
# Ejercicio 16.03 Crea un conjunto por comprensión que normalice etiquetas y un diccionario por comprensión que asocie cada etiqueta con su longitud. Explica qué sucede si dos entradas normalizan igual.

etiquetas = [" aula_01 ", "AULA-01", "aula_03"]
etiquetas_normalizadas = {etiqueta.strip().lower().replace("_", "-") for etiqueta in etiquetas}
diccionario_compresion = {etiqueta: len(etiqueta) for etiqueta in etiquetas_normalizadas}

# Pruebas
assert etiquetas_normalizadas == {"aula-01","aula-03"}
assert diccionario_compresion == {"aula-01": 7, "aula-03": 7}

# Si hay dos entradas duplicadas el último valor remplazaria a la que ya existe dado que en un diccionario no pueden haber valores repetidos con la misma clave
```

## Proyecto 16 · Procesador perezoso de lecturas

Implementa un generador lecturas_validas que reciba textos de temperatura y produzca solo los valores convertibles a float, omitiendo los inválidos y registrando cuántos se descartaron mediante una estructura acordada. Crea comprensiones para clasificar las lecturas adecuadas y generar mensajes, y compara mediante Markdown qué resultado conservarías como lista y cuál como generador.

Incluye pruebas para texto válido, inválido, límite 18, límite 27 y generador agotado. Explica por qué una comprensión puede ser menos apropiada que un bucle si se necesita registrar cada error descartado.

```python
def lecturas_validad(textos: list[str], contador: dict):
    """ Generador que intenta convertir textos a float. Omite los inválidos y aumenta el contador de 'descartados'"""
    for texto in textos:
        try:
            yield float(texto)
        except ValueError:
            contador["descartados"] += 1

textos_prueba = ["18.0","22.0","veinte","27.0"]
registro_descartados = {"descartados": 0}
generador = lecturas_validad(textos_prueba, registro_descartados)
lecturas_adecuadas = [lectura for lectura in generador if 18 <= lectura <= 27]
mensajes_compresion = [f"{temperatura:.1f} ºC" for temperatura in lecturas_adecuadas]

# Pruebas
assert lecturas_adecuadas == [18.0, 22.0, 27.0]
assert mensajes_compresion == ["18.0 ºC", "22.0 ºC", "27.0 ºC"]
assert registro_descartados["descartados"] == 1
assert list(generador) == []

```
