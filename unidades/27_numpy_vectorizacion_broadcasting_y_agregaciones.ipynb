## 1. Operaciones vectorizadas y `ufuncs`

Una expresión como `temperaturas + 273.15` se aplica elemento a elemento. Funciones como `np.sqrt`, `np.log`, `np.clip` y `np.isfinite` trabajan sobre arrays completos. Antes de usarlas, revisa su dominio: por ejemplo, el logaritmo de un valor no positivo no es un resultado válido.

```python
temperaturas = np.array([20.5, 22.3, 21.1, 23.0])
kelvin = temperaturas + 273.15
centradas = temperaturas - temperaturas.mean()
print(kelvin)
print(centradas)
```

## 2. Broadcasting

Dos formas son compatibles si, al compararlas desde la derecha, coinciden o una de las dimensiones vale 1. Una matriz `(2, 3)` puede sumar un vector `(3,)`: el vector se aplica a cada fila. No cambies formas al azar para eliminar un error; primero identifica qué representa cada eje.

```python
medidas = np.array([[10, 20, 30], [12, 22, 32]], dtype=float)
correccion = np.array([0.5, -1.0, 0.0])
corregidas = medidas + correccion
print(corregidas)
```

## 3. Agregar por ejes

Sin `axis`, una agregación resume todo el array. En una matriz con filas como días y columnas como sensores, `axis=0` da un valor por sensor y `axis=1` da un valor por día. Elige el eje a partir de su significado, no de un resultado que parezca razonable.

```python
lecturas = np.array([[12, 15, 14], [13, 16, 15], [11, 14, 13]])
print("Media por sensor:", lecturas.mean(axis=0))
print("Media por día:", lecturas.mean(axis=1))
```

## 4. Condiciones y ausencias numéricas

`np.where` escoge valores según una condición. `np.isnan` detecta ausencias numéricas. Una normalización min–max exige que máximo y mínimo sean distintos. Ante un valor ausente debes decidir si excluir, imputar o detener el análisis; no conviene convertirlo en cero sin justificarlo.

```python
probabilidades = np.array([0.2, 0.65, 0.9, 0.4, 0.78])
clases = np.where(probabilidades >= 0.5, 1, 0)
print(clases)
print("Positivos:", clases.sum())
```

```python
# 27.01. Convierte temperaturas a Kelvin y Fahrenheit sin bucles, y calcula su media.
celsius = np.array([20.5, 22.3, 21.1, 23.0, 19.8])

kelvin = celsius + 273.15
fahrenheit = celsius * 1.8 + 32

media_kelvin = kelvin.mean() # media aritmetica de las temperaturas en Kelvin
media_fahrenheit = fahrenheit.mean() # media aritmetica de las temperaturas en Fahrenheit

print("Celsius:   ", celsius, f"| Media: {celsius.mean():.2f} °C")
print("Kelvin:    ", kelvin, f"| Media: {media_kelvin:.2f} K")
print("Fahrenheit:", fahrenheit, f"| Media: {media_fahrenheit:.2f} °F")
```

```python
# 27.02. Normaliza min–max un array de temperaturas. Indica qué harías si todos los valores fueran iguales.

temperaturas = np.array([18.0, 21.5, 20.0, 23.5])

minimo = temperaturas.min()
maximo = temperaturas.max()
rango = maximo - minimo

if rango != 0:
    normalizadas = (temperaturas - minimo) / rango
else:
    normalizadas = np.zeros_like(temperaturas)

print("Temperaturas originales:", temperaturas)
print("Temperaturas normalizadas [0, 1]:", normalizadas)

# Si todos los valores fueran iguales, max() - min() sería 0, lo que causaría una división por cero. Por
# eso de comprueba primero que max != min y si son iguales, se puede devolver un array de ceros en lugar de dividir.
```

```python
# 27.03. Para una matriz 3×3, calcula suma por filas y desviación típica por columnas, explicando los ejes.
matriz = np.array(
    [[10, 20, 30],
     [12, 25, 33],
     [14, 15, 27]])
suma_filas = matriz.sum(axis=1)  # Suma a lo largo de las filas (axis=1)

desviacion_columnas = matriz.std(axis=0)  # Desviación estándar a lo largo de las columnas (axis=0)

print("Matriz:\n", matriz)
print("\nSuma por filas (axis=1):", suma_filas)
print("Desviación típica por columnas (axis=0):", desviacion_columnas)
```

```python
# 27.04. Clasifica probabilidades como 1 si son al menos 0,5 y 0 en otro caso; informa de media y positivos.
probabilidades = np.array([0.2, 0.65, 0.9, 0.4, 0.78, 0.5])

# 2. Clasificación vectorizada: 1 si >= 0.5, 0 en caso contrario
clases = np.where(probabilidades >= 0.5, 1, 0)

# 3. Agregaciones sobre las clases (o sobre probabilidades)
positivos = clases.sum()
media_clases = clases.mean()           # Proporción de positivos (4 / 6 = 0.6667)
media_probs = probabilidades.mean()    # Media de las probabilidades originales

print("Probabilidades:", probabilidades)
print("Clases (umbral 0.5):", clases)
print("Total positivos:", positivos)
print(f"Media de clases (proporción de positivos): {media_clases:.2f}")
print(f"Media de probabilidades: {media_probs:.2f}")
```

```python
# 27.05. Aplica una corrección distinta a cada columna de una matriz `(4, 3)` mediante broadcasting y verifica la forma.

import numpy as np

# 1. Creamos una matriz de forma (4, 3) (4 observaciones x 3 variables/columnas)
matriz = np.array([
    [10.0, 20.0, 30.0],
    [12.0, 22.0, 32.0],
    [14.0, 24.0, 34.0],
    [16.0, 26.0, 36.0]
])

# 2. Vector de corrección con 3 valores distintos (uno por columna), forma (3,)
correccion = np.array([0.5, -1.0, 2.0])

# 3. Aplicamos la corrección mediante broadcasting
corregida = matriz + correccion

# 4. Verificamos las formas (.shape)
print("Forma matriz original:", matriz.shape)
print("Forma vector corrección:", correccion.shape)
print("Forma matriz corregida:", corregida.shape)
print("\nMatriz corregida:\n", corregida)

# Comprobación explícita de que la forma se mantiene en (4, 3)
assert corregida.shape == (4, 3)
```
