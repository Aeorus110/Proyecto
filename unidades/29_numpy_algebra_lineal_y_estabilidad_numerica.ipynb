## 1. Dos productos diferentes

`A * B` multiplica valores que comparten posición. `A @ B` compone transformaciones: las columnas de `A` deben coincidir con las filas de `B`. Ambas operaciones pueden devolver una matriz de la misma forma, pero significan cosas distintas. Comprueba siempre las formas antes de calcular.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("Elemento a elemento:")
print(a * b)
print("Matricial:")
print(a @ b)
```

## 2. Transpuesta e identidad

La transpuesta intercambia filas y columnas. La identidad es una matriz cuadrada que no cambia otra matriz compatible al multiplicarla. Estas propiedades permiten comprobar resultados, pero una identidad no puede arreglar una incompatibilidad de formas.

```python
c = np.arange(1, 10).reshape(3, 3)
identidad = np.eye(3, dtype=int)
print(c.T)
print(c @ identidad)
```

## 3. Resolver sistemas

Para resolver `A @ x = b`, usa `np.linalg.solve(A, b)` cuando `A` es cuadrada y existe una solución única. Evita calcular una inversa sólo para multiplicarla por `b`: es menos claro y puede ser menos estable. Si el sistema es singular, `solve` debe avisarte y la situación necesita otra formulación.

```python
a = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([5.0, 6.0])
x = np.linalg.solve(a, b)
print(x)
print(a @ x)
```

## 4. Normas y coma flotante

Una norma resume la magnitud de un vector. La norma 2 se relaciona con distancia euclidiana; la norma 1 suma valores absolutos. Los decimales se almacenan con precisión finita: para comprobar cálculos usa `np.allclose`, no igualdad exacta. El número de condición alerta de sistemas sensibles a pequeños cambios.

```python
vector = np.array([3.0, 4.0, 0.0])
print(np.linalg.norm(vector, ord=1))
print(np.linalg.norm(vector, ord=2))
print(np.allclose(0.1 + 0.2, 0.3))
```

## Práctica guiada

Crea dos matrices 2×2. Calcula `*` y `@`; crea una identidad compatible; y verifica que multiplicar por ella conserva la matriz. Después resuelve un sistema 2×2 y comprueba que `A @ x` coincide aproximadamente con `b`.

```python
a = np.array([[1.0, 1.0], [2.0, 3.0]])
b = np.array([4.0, 9.0])
solucion = np.linalg.solve(a, b)
assert np.allclose(a @ solucion, b)
print(solucion)
```
