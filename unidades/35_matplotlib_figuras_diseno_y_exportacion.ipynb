```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 1. Una figura, varios ejes

`fig, ejes = plt.subplots(filas, columnas, ...)` crea una figura y una cuadrícula de ejes. Si hay más de un eje, `ejes` es normalmente un array de objetos `Axes`; por eso puedes acceder a cada panel con índices como `ejes[0, 1]`.

La figura debe tener un propósito común. Reunir gráficos sin relación solo añade carga visual. En el ejemplo, los tres paneles describen la misma serie de lecturas desde perspectivas distintas: evolución, distribución y resumen por grupo.

```python
horas = np.array([8, 10, 12, 14, 16, 18])
temperatura_c = np.array([14.2, 16.8, 19.1, 20.0, 18.4, 16.1])
turno = np.array(["Mañana", "Mañana", "Mañana", "Tarde", "Tarde", "Tarde"])

fig, ejes = plt.subplots(1, 3, figsize=(12, 3.8), layout="constrained")
ejes[0].plot(horas, temperatura_c, marker="o")
ejes[0].set(title="Evolución", xlabel="Hora", ylabel="Temperatura (°C)")
ejes[0].set_xticks(horas)
ejes[0].grid(axis="y", alpha=0.25)

ejes[1].hist(temperatura_c, bins=4, color="tab:orange", edgecolor="white")
ejes[1].set(title="Distribución", xlabel="Temperatura (°C)", ylabel="Lecturas")
ejes[1].grid(axis="y", alpha=0.25)

resumen = pd.DataFrame({"turno": turno, "temperatura_c": temperatura_c}).groupby("turno", as_index=False)["temperatura_c"].mean()
barras = ejes[2].bar(resumen["turno"], resumen["temperatura_c"], color=["tab:blue", "tab:green"])
ejes[2].set(title="Media por turno", xlabel="Turno", ylabel="Temperatura media (°C)")
ejes[2].bar_label(barras, fmt="%.1f", padding=2)
ejes[2].grid(axis="y", alpha=0.25)

fig.suptitle("Lecturas de temperatura en una jornada")
plt.show()
```

## 2. Diseño que ayuda a leer

Una figura legible tiene jerarquía: un título global da contexto, cada eje explica su medida y las leyendas solo aparecen cuando identifican series. `layout="constrained"` pide a Matplotlib que distribuya automáticamente etiquetas, títulos y ejes. Para figuras creadas sin ese argumento, `fig.tight_layout()` puede resolver muchos solapamientos.

Evita recargar: no necesitas cuadrícula, leyenda, anotación y colores intensos en todos los paneles. Cada elemento debe responder a una necesidad de lectura. Mantén unidades y escalas visibles; no dependas solo del color para diferenciar categorías.

## 3. Anotar un dato no explica su causa

`ax.annotate` asocia un texto a una coordenada. Es adecuado para destacar un máximo, un valor atípico o una fecha relevante. La anotación debe describir lo observado; si propone una hipótesis, indícala como tal y no como un hecho demostrado.

```python
dias = np.arange(1, 8)
incidencias = np.array([4, 5, 3, 9, 6, 4, 5])
indice_maximo = incidencias.argmax()

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(dias, incidencias, marker="o", label="Incidencias registradas")
ax.annotate(
    "Máximo observado: 9",
    xy=(dias[indice_maximo], incidencias[indice_maximo]),
    xytext=(dias[indice_maximo] + 0.5, incidencias[indice_maximo] + 1.2),
    arrowprops={"arrowstyle": "->"},
)
ax.set(title="Incidencias por día", xlabel="Día", ylabel="Número de incidencias")
ax.set_xticks(dias)
ax.grid(axis="y", alpha=0.25)
ax.legend()
plt.show()
```

## Lista de revisión de accesibilidad

Antes de entregar una figura, comprueba:

1. título que aporta contexto y ejes con nombre y unidad;
2. tamaño de figura y fuentes legibles al verla fuera del cuaderno;
3. contraste suficiente entre elementos y fondo;
4. categorías distinguibles por leyenda, marcador o patrón además del color;
5. explicación textual de la figura en el informe, para quien no pueda verla o la consulte impresa.

## Práctica guiada

Crea un panel de dos gráficos con los datos de mantenimiento siguientes. El primer eje debe mostrar la evolución temporal; el segundo, la distribución. Luego:

1. usa un título global que relacione ambos paneles;
2. añade una anotación en el máximo;
3. revisa qué información se repite y elimina la que no aporte;
4. escribe una frase que describa el máximo sin asignarle una causa.

```python
semanas = np.arange(1, 7)
horas_mantenimiento = np.array([8, 11, 9, 15, 12, 10])
maximo = horas_mantenimiento.argmax()

fig, ejes = plt.subplots(1, 2, figsize=(9, 4), layout="constrained")
ejes[0].plot(semanas, horas_mantenimiento, marker="o", label="Horas")
ejes[0].annotate("Máximo", xy=(semanas[maximo], horas_mantenimiento[maximo]), xytext=(4.3, 16), arrowprops={"arrowstyle": "->"})
ejes[0].set(title="Evolución semanal", xlabel="Semana", ylabel="Horas de mantenimiento")
ejes[0].grid(axis="y", alpha=0.25)
ejes[0].legend()
ejes[1].hist(horas_mantenimiento, bins=4, color="tab:orange", edgecolor="white")
ejes[1].set(title="Distribución de horas", xlabel="Horas de mantenimiento", ylabel="Semanas")
ejes[1].grid(axis="y", alpha=0.25)
fig.suptitle("Mantenimiento durante seis semanas")
plt.show()
```
