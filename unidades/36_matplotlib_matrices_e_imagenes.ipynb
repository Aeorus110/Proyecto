```python
import matplotlib.pyplot as plt
import numpy as np
```

## 1. De una matriz a una imagen

Para una matriz bidimensional, cada celda tiene una posición de fila y columna y un valor. `imshow` asigna colores a esos valores. Por defecto, la primera fila aparece arriba; `origin="lower"` sitúa la primera fila abajo, lo cual puede ser más intuitivo cuando las filas representan coordenadas verticales crecientes.

El mapa de color no cambia los datos. `cmap="viridis"` es una opción legible para valores ordenados; evita mapas que cambien de luminosidad de forma irregular cuando la intensidad es importante. La barra de color explica qué valor corresponde a cada tono.

```python
temperatura_c = np.array(
    [
        [16.1, 16.8, 17.5, 18.0],
        [15.8, 16.6, 17.3, 17.9],
        [15.2, 16.0, 16.9, 17.4],
    ]
)

fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
imagen = ax.imshow(temperatura_c, cmap="viridis", vmin=15, vmax=19, origin="lower")
barra = fig.colorbar(imagen, ax=ax, label="Temperatura (°C)")
ax.set(title="Temperatura en una cuadrícula sintética", xlabel="Columna", ylabel="Fila")
ax.set_xticks(np.arange(temperatura_c.shape[1]))
ax.set_yticks(np.arange(temperatura_c.shape[0]))
plt.show()

print("Forma del array:", temperatura_c.shape, "(filas, columnas)")
```

## 2. Escala y comparación honesta

Los límites `vmin` y `vmax` fijan la correspondencia entre valor y color. Si comparas dos matrices, usa los mismos límites y el mismo mapa de color; de lo contrario, el mismo tono puede representar valores distintos en cada panel.

Una celda más clara o más oscura no es «mejor» o «peor» por sí misma. La interpretación depende de la variable y de la escala. Escribe la unidad en la barra de color y el contexto en el título.

```python
matriz_a = np.array([[2, 4, 6], [3, 5, 7], [4, 6, 8]])
matriz_b = np.array([[1, 3, 5], [2, 6, 9], [4, 7, 10]])

fig, ejes = plt.subplots(1, 2, figsize=(9, 4), layout="constrained")
for ax, matriz, titulo in zip(ejes, [matriz_a, matriz_b], ["Zona A", "Zona B"]):
    imagen = ax.imshow(matriz, cmap="magma", vmin=0, vmax=10, origin="lower")
    ax.set(title=titulo, xlabel="Columna", ylabel="Fila")
fig.colorbar(imagen, ax=ejes, label="Conteo (casos)")
fig.suptitle("Dos matrices con una escala compartida")
plt.show()
```

## 3. Matrices de conteos y etiquetas

Una matriz de conteos puede resumir pares de categorías. Las filas y columnas deben quedar etiquetadas; sin ellas, el color no permite saber qué se compara. Para matrices pequeñas, escribir el valor en cada celda evita exigir que la persona lectora estime números por el tono.

Este patrón se reutilizará en la unidad 37 al comunicar resultados precomputados. La matriz es un resumen de datos de entrada, no una explicación de por qué se producen los errores.

```python
conteos = np.array([[18, 2, 1], [3, 14, 2], [1, 2, 17]])
etiquetas = ["A", "B", "C"]

fig, ax = plt.subplots(figsize=(5, 4), layout="constrained")
imagen = ax.imshow(conteos, cmap="Blues", vmin=0, vmax=20)
fig.colorbar(imagen, ax=ax, label="Número de casos")
ax.set(title="Matriz de conteos", xlabel="Categoría de columna", ylabel="Categoría de fila")
ax.set_xticks(np.arange(len(etiquetas)), etiquetas)
ax.set_yticks(np.arange(len(etiquetas)), etiquetas)
for fila in range(conteos.shape[0]):
    for columna in range(conteos.shape[1]):
        color_texto = "white" if conteos[fila, columna] > 10 else "black"
        ax.text(columna, fila, str(conteos[fila, columna]), ha="center", va="center", color=color_texto)
plt.show()
```

## 4. Imágenes en escala de grises y color

- Una imagen en escala de grises suele tener forma `(alto, ancho)`: un valor por píxel.
- Una imagen RGB suele tener forma `(alto, ancho, 3)`: tres canales por píxel.

En el ejemplo no cargamos una fotografía externa: construimos una imagen sintética a partir de arrays. Esto permite comprobar la forma y entender que `imshow` no reconoce objetos ni realiza ninguna predicción.

```python
imagen_rgb = np.zeros((8, 8, 3), dtype=float)
imagen_rgb[:, :4, 0] = 0.85  # Mitad izquierda: componente roja.
imagen_rgb[:, 4:, 2] = 0.85  # Mitad derecha: componente azul.
imagen_rgb[2:6, 2:6, 1] = 0.55  # Zona central: componente verde.

fig, ax = plt.subplots(figsize=(4, 4))
ax.imshow(imagen_rgb, interpolation="nearest")
ax.set(title="Imagen RGB sintética", xlabel="Columna de píxel", ylabel="Fila de píxel")
ax.set_xticks(np.arange(0, 8, 2))
ax.set_yticks(np.arange(0, 8, 2))
plt.show()
print("Forma RGB:", imagen_rgb.shape, "(alto, ancho, canales)")
```

## Práctica guiada

La matriz siguiente contiene valores simulados de ocupación. Ejecuta el código y después:

1. identifica la fila y columna de mayor valor con `np.unravel_index`;
2. cambia el mapa de color y comprueba que los valores no cambian;
3. añade el valor máximo como anotación textual;
4. escribe una frase que describa el patrón sin explicar su causa.

```python
ocupacion = np.array([[2, 4, 3, 1], [5, 8, 6, 2], [3, 7, 9, 4]])
fila, columna = np.unravel_index(ocupacion.argmax(), ocupacion.shape)

fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
imagen = ax.imshow(ocupacion, cmap="cividis", vmin=0, vmax=10, origin="lower")
fig.colorbar(imagen, ax=ax, label="Ocupación (personas)")
ax.set(title="Ocupación simulada por celda", xlabel="Columna", ylabel="Fila")
ax.scatter(columna, fila, facecolors="none", edgecolors="white", s=180, linewidths=2, label="Máximo")
ax.legend()
plt.show()
print("Máximo:", ocupacion[fila, columna], "en fila", fila, "y columna", columna)
```
