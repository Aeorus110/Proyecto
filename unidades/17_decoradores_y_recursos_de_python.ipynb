## 1. Las funciones son objetos

En Python una función se puede guardar en una variable, pasar como argumento y devolver desde otra función. Escribe el nombre sin paréntesis para referirte a la función; usa paréntesis para llamarla. Esta propiedad permite construir herramientas que reciben un comportamiento, como sorted con una función clave, o decoradores que amplían otra función.

No es un recurso para complicar programas pequeños. Es útil cuando varias funciones deben recibir el mismo tratamiento, por ejemplo registrar una llamada, medir una duración o comprobar una condición común.

```python
def duplicar(numero: int) -> int:
    return numero * 2

operacion = duplicar
print(operacion(4))

def aplicar(valor: int, funcion) -> int:
    return funcion(valor)

print(aplicar(5, duplicar))
```

## 2. Funciones internas y cierres

Una función puede definirse dentro de otra. La función interna puede recordar valores del ámbito exterior incluso después de que termine la llamada exterior; esto se denomina cierre. Un cierre es la base conceptual de muchos decoradores.

Úsalo cuando necesites configurar un comportamiento y devolver una función resultante. Mantén los cierres cortos y con nombres claros. Si el estado debe ser complejo o modificable, una clase puede expresar mejor el diseño.

```python
def crear_saludo(prefijo: str):
    def saludar(nombre: str) -> str:
        return f"{prefijo}, {nombre}"
    return saludar

saludo_formal = crear_saludo("Buenos días")
print(saludo_formal("Ana"))
```

## 3. Qué hace un decorador

Un decorador recibe una función y devuelve otra función que añade comportamiento antes o después de la llamada original. La sintaxis con arroba encima de una definición es una forma abreviada de reemplazar la función por el resultado del decorador.

Un decorador útil debe respetar el contrato de la función decorada: aceptar los argumentos necesarios, devolver su resultado y no cambiar el significado sin documentarlo. Si solo se utiliza una vez y dificulta la lectura, una llamada explícita puede ser preferible.

```python
def registrar_llamada(funcion):
    def envoltura(*args, **kwargs):
        print("Inicio:", funcion.__name__)
        resultado = funcion(*args, **kwargs)
        print("Fin:", funcion.__name__)
        return resultado
    return envoltura

@registrar_llamada
def clasificar(temperatura: float) -> str:
    return "adecuada" if 18 <= temperatura <= 27 else "revisar"

print(clasificar(22.0))
```

### Metadatos y functools.wraps

La función envoltura sustituye a la original y puede ocultar su nombre, docstring y otras propiedades. Functools.wraps copia esos metadatos relevantes. Es una práctica recomendable en decoradores reutilizables porque mejora la depuración, la documentación y algunas herramientas de prueba.

Args recoge argumentos posicionales y kwargs argumentos por nombre. Permiten que un decorador sencillo funcione con distintas firmas, pero no sustituyen a entender qué argumentos espera la función original.

```python
from functools import wraps

def registrar(funcion):
    @wraps(funcion)
    def envoltura(*args, **kwargs):
        print(f"Llamando a {funcion.__name__}")
        return funcion(*args, **kwargs)
    return envoltura

@registrar
def convertir(texto: str) -> float:
    """Convierte una lectura de texto a número decimal."""
    return float(texto)

print(convertir.__name__)
print(convertir("21.4"))
```

## 4. Anotaciones y código actual

Las anotaciones documentan tipos esperados y retornos. En Python moderno se pueden usar formas como list[str], dict[str, float] y float | None. Ayudan a lectores, editores y analizadores estáticos, pero no validan valores automáticamente durante la ejecución.

No añadas anotaciones solo para parecer avanzado. Deben mejorar la lectura del contrato. Mantén las funciones pequeñas, usa docstrings cuando la intención no sea evidente y valida datos externos de forma explícita.

**Ejercicio 17.01.** Crea aplicar_a_todos que reciba una lista y una función, y devuelva resultados aplicados. Pruébala con cuadrado y con convertir texto a mayúsculas.

**Ejercicio 17.02.** Crea un decorador avisar_errores que muestre un mensaje antes de propagar ValueError de una función de conversión. Usa wraps.

**Ejercicio 17.03.** Añade anotaciones y docstrings a una función media_segura que devuelva None para una lista vacía. Escribe pruebas de ambos casos.

```python
# Ejercicio 17.01. Crea aplicar_a_todos que reciba una lista y una función, y devuelva resultados aplicados. Pruébala con cuadrado y con convertir texto a mayúsculas.
def aplicar_a_todos(elementos: list, funcion) -> list:
    return [funcion(elemento) for elemento in elementos]

def cuadrado(numero:int) -> int:
    return numero ** 2

def a_mayusculas(texto: str) -> str:
    return texto.upper()

# Pruebas
numeros = [1,2,3]
textos = ["aula", "humedad"]

assert aplicar_a_todos(numeros, cuadrado) == [1,4,9]
assert aplicar_a_todos(textos, a_mayusculas) == ["AULA", "HUMEDAD"]
```

```python
# Ejercicio 17.02. Crea un decorador avisar_errores que muestre un mensaje antes de propagar ValueError de una función de conversión. Usa wraps.
from functools import wraps

def avisar_errores(funcion):
    @wraps(funcion)
    def envoltura(*args, **kwargs):
        try:
            return funcion(*args, **kwargs)
        except ValueError as error:
            print(f"La función {funcion.__name__} falló: {error}")
            raise
    return envoltura

@avisar_errores
def convertir_humedad(texto: str) -> int:
    return int(texto)

# Pruebas
assert convertir_humedad.__name__ == "convertir_humedad"
assert convertir_humedad("45") == 45

try:
    convertir_humedad("veinte")
    assert False
except ValueError:
    pass
```

```python
# Ejercicio 17.03. Añade anotaciones y docstrings a una función media_segura que devuelva None para una lista vacía. Escribe pruebas de ambos casos.
def media(numeros: list[float]) -> float | None:
    """Calcula la media de la lista de números"""
    if not numeros:
        return None
    return sum(numeros) / len(numeros)

# Pruebas
assert media([]) is None
assert media([1,5,9]) == 5
```

## Proyecto 17 · Pipeline de lecturas auditable

Construye un pequeño pipeline con funciones convertir_lectura, clasificar_lectura y resumir_lecturas. Añade un decorador registrar_operacion que conserve metadatos y muestre la función llamada. El pipeline debe convertir textos, separar valores inválidos, clasificar los válidos y devolver un informe con anotaciones y docstrings.

Incluye pruebas de conversión válida e inválida, límites 18 y 27, lista vacía y presencia del decorador. Documenta qué parte añade el decorador, por qué no modifica la regla de clasificación y qué revisarías antes de usar una biblioteca externa en un proyecto de IA.

```python
from functools import wraps

def registrar_operacion(funcion):
    """Imprime el nombre de la función que se está ejecutando"""
    @wraps(funcion)
    def envoltura(*args, **kwargs):
        print(f"Función: {funcion.__name__}")
        return funcion(*args, **kwargs)
    return envoltura

@registrar_operacion
def convertir_lectura(texto: str) -> float:
    """Convierte un string a decimal"""
    return float(texto)

@registrar_operacion
def clasificar_lectura(temperatura: float) -> str:
    """Clasifica la temperatura como adecuada o revisar"""
    if 18 <= temperatura <= 27:
        return "adecuada"
    return "revisar"

@registrar_operacion
def resumir_lecturas(textos: list[str]) -> dict:
    """Recibe una lista de textos y devuelve un informe clasificado"""
    informe = {"adecuada": [], "revisar": [], "errores_conversion": 0}

    for texto in textos:
        try:
            temperatura = convertir_lectura(texto)
            clasificacion = clasificar_lectura(temperatura)
            informe[clasificacion].append(temperatura)
        except ValueError:
            informe["errores_conversion"] += 1

    return informe

# Pruebas
textos = ["21.5", "18.0", "cuatro", "27.0", "30.5"]
resultado = resumir_lecturas(textos)

assert resultado["adecuada"] == [21.5, 18.0,27.0]
assert resultado["revisar"] == [30.5]
assert resultado["errores_conversion"] == 1

resultado_vacio = resumir_lecturas([])
assert resultado_vacio["adecuada"] == []
assert resultado_vacio["errores_conversion"] == 0

assert resumir_lecturas.__name__ == "resumir_lecturas"
```
