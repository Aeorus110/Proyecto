## 1. Pregunta antes que gráfico

Una barra compara categorías, una línea muestra evolución temporal ordenada, un histograma describe una distribución, una dispersión explora relación entre dos variables numéricas y un diagrama de cajas compara distribuciones y valores extremos. Elegir por costumbre puede ocultar la pregunta.

Antes de representar, declara población, unidad, variable y periodo. Este CSV es didáctico y pequeño: permite practicar código, no describir una ciudad real ni demostrar causalidad.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ruta = Path('..') / 'datos' / 'pandas' / '20_lecturas_estaciones_limpias.csv'
if not ruta.exists():
    ruta = Path('datos') / 'pandas' / '20_lecturas_estaciones_limpias.csv'
lecturas = pd.read_csv(ruta)
lecturas['momento'] = pd.to_datetime(lecturas['fecha'] + ' ' + lecturas['hora'])
lecturas.head()
```

## 2. Barras: comparar una medida agregada

No dibujes una barra por cada lectura si la pregunta es comparar zonas. Primero agrupa, nombra la medida y después representa. El eje vertical debe indicar unidad y el título debe describir qué se ha calculado.

```python
pm25_zona = lecturas.groupby('zona', as_index=False)['pm25_ug_m3'].mean()
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(pm25_zona['zona'], pm25_zona['pm25_ug_m3'], color=['#4C78A8', '#F58518', '#54A24B'])
ax.set(title='Media de PM2.5 por zona', xlabel='Zona', ylabel='PM2.5 media (µg/m³)')
ax.grid(axis='y', alpha=0.25)
plt.show()
```

## 3. Distribución y relación

El histograma agrupa valores en intervalos: cambiar el número de barras puede cambiar la apariencia, por lo que debe elegirse con criterio. La dispersión coloca cada observación en dos ejes; una forma ascendente o descendente sugiere asociación, nunca causalidad.

Color, texto alternativo, etiquetas y símbolos no deben ser los únicos elementos que comuniquen significado. Evita gráficos de pastel cuando comparar longitudes sea más preciso.

```python
fig, ejes = plt.subplots(1, 2, figsize=(11, 4))
ejes[0].hist(lecturas['pm25_ug_m3'], bins=6, color='#4C78A8', edgecolor='white')
ejes[0].set(title='Distribución de PM2.5', xlabel='PM2.5 (µg/m³)', ylabel='Número de lecturas')
ejes[1].scatter(lecturas['temperatura_c'], lecturas['pm25_ug_m3'], color='#F58518')
ejes[1].set(title='Temperatura y PM2.5', xlabel='Temperatura (°C)', ylabel='PM2.5 (µg/m³)')
for eje in ejes: eje.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

## 4. Tiempo y diagramas de cajas

Una línea exige ordenar el eje temporal. Un boxplot resume mediana, cuartiles y posibles valores alejados, pero no muestra cada detalle de la distribución. Acompaña siempre el gráfico de una interpretación prudente y, cuando sea posible, del número de observaciones por grupo.

```python
serie = lecturas.sort_values('momento')
fig, ejes = plt.subplots(1, 2, figsize=(12, 4))
ejes[0].plot(serie['momento'], serie['pm25_ug_m3'], marker='o')
ejes[0].set(title='PM2.5 en el tiempo', xlabel='Momento', ylabel='PM2.5 (µg/m³)')
ejes[0].tick_params(axis='x', rotation=45)
lecturas.boxplot(column='pm25_ug_m3', by='zona', ax=ejes[1])
ejes[1].set(title='Distribución de PM2.5 por zona', xlabel='Zona', ylabel='PM2.5 (µg/m³)')
plt.suptitle('')
plt.tight_layout()
plt.show()
```

## Errores frecuentes

- Representar datos sin formular una pregunta.
- Usar escalas truncadas sin advertirlo.
- Omitir unidades y fuente de los datos.
- Inferir causalidad de una dispersión.
- Usar demasiados colores o depender solo de ellos.
- Exportar una figura sin revisar su tamaño, etiquetas y legibilidad.
