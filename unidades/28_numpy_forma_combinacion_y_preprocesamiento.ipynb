## 1. Cambiar forma sin cambiar datos

`reshape` sólo funciona si el número total de elementos coincide. `-1` permite que NumPy calcule una dimensión, pero sólo debe usarse si puedes explicar qué tamaño resultará. `ravel` intenta devolver una vista aplanada; `flatten` crea una copia. Esa diferencia importa si modificarás el resultado.

```python
valores = np.arange(12)
matriz = valores.reshape(3, 4)
print(matriz)
print(matriz.reshape(2, -1))
print(matriz.ravel().shape)
```

## 2. Transponer, concatenar y dividir

`.T` intercambia ejes en una matriz 2D. `np.vstack` agrega filas; `np.hstack` agrega columnas; `np.concatenate` permite indicar el eje. Las formas compatibles no garantizan unidades compatibles: antes de combinar, verifica que cada columna representa la misma variable y escala.

```python
manana = np.array([[12, 60], [14, 55]])
tarde = np.array([[20, 45], [22, 40]])
dia = np.vstack([manana, tarde])
print(dia)
print(dia.T)
```

## 3. Preparación numérica

En muchas tareas posteriores, las filas son observaciones y las columnas son variables. Si la última columna es la variable que quieres estudiar, puedes separar `X` e `y`. Esta separación no crea un modelo ni convierte una asociación en una predicción: sólo organiza los datos para el siguiente paso.

```python
datos = np.array([[12.0, 60.0, 18.0], [14.0, 55.0, 15.0], [20.0, 45.0, 22.0]])
x = datos[:, :2]
y = datos[:, 2]
print(x.shape, y.shape)
```

```python
# 28.01 Reorganiza 24 valores en una matriz 6×4, extrae las tres primeras filas 
# y divídela en dos bloques de 3×4.
matriz = np.arange(24).reshape(6, 4) # Reshape se usa para converti la matriz en 2D.

tres_primeras_filas = matriz[:3]

bloque1, bloque2 = np.vsplit(matriz, 2) # Divide la matriz.

print("Tres primeras filas:\n", tres_primeras_filas)
print("\nBloque 1:\n", bloque1)
print("\nBloque 2:\n", bloque2)
```

```python
# 28.02 Aplana una matriz 3×4 con `ravel` y `flatten`. 
# Explica cuál es más seguro si modificarás el resultado.

matriz = np.arange(12).reshape(3,4)

vista_plana_ravel = matriz.ravel()
copia_plana = matriz.flatten()

vista_plana_ravel[0] = 999
copia_plana[1] = 888

print("Matriz original tras modificar ambos:\n", matriz)

# Explicación: Es más seguro usar flatten() si vas a modificar el resultado 
# porque siempre devuelve una copia independiente en memoria; en cambio, ravel() 
# intenta devolver una vista que comparte memoria con el array original, 
# por lo que editarlo puede alterar tus datos originales sin querer.
```

```python
# 28.03 Concatena dos matrices 2×3 vertical y horizontalmente. Indica las formas de cada resultado.

matriz1 = np.arange(6).reshape(2,3)
matriz2 = np.arange(6,12).reshape(2,3)

vertical = np.vstack([matriz1, matriz2]) # (4,3)

horizontal = np.hstack([matriz1, matriz2]) # (2,6)

print("Vertical:\n", vertical)
print("Forma vertical:", vertical.shape)

print("\nHorizontal:\n", horizontal)
print("Forma horizontal:", horizontal.shape)
```

```python
# **28.04.** Simula un lote de 5 imágenes 2×2 y transfórmalo a `(5, 4)`. 
# Explica qué conserva cada fila.

lote = np.arange(20).reshape(5, 2, 2)
lote_plano = lote.reshape(5,4)

print("Forma lote original:", lote.shape)
print("Forma lote aplanado:", lote_plano.shape)
print("\nLote (5, 4):\n", lote_plano)

# Explicación: Cada fila conserva una imagen completa del lote (una observación) 
# con sus 4 píxeles ordenados fila por fila, manteniendo separadas las distintas imágenes.
```
