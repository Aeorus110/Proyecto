## 1. Qué es una cadena

Una cadena, de tipo str, es una secuencia ordenada de caracteres. Se escribe entre comillas simples o dobles. La elección entre unas y otras es una cuestión de estilo; puede ser útil elegir el tipo que evite tener que escapar una comilla incluida en el texto.

Las cadenas son **inmutables**. Una operación que parece modificar texto, como convertirlo a minúsculas, crea una nueva cadena y deja intacta la original. Si quieres conservar el resultado transformado debes asignarlo a otro nombre o reasignarlo explícitamente al mismo. Esta propiedad evita cambios inesperados cuando varias partes del programa usan el mismo texto.

Las cadenas admiten longitud, índices y cortes igual que las listas. El primer carácter tiene índice 0, el último índice puede escribirse como menos uno y los cortes excluyen el extremo final. Pedir un índice inexistente genera IndexError.

```python
mensaje = "Datos para IA"
print(mensaje)
print("Longitud:", len(mensaje))
print("Primer carácter:", mensaje[0])
print("Último carácter:", mensaje[-1])
print("Corte:", mensaje[6:9])
```

### Comillas, saltos y representación

Algunos caracteres tienen una función especial dentro de una cadena. La barra invertida permite escribir secuencias de escape: barra invertida n representa un salto de línea, barra invertida t una tabulación y una barra invertida antes de comillas permite incluirlas en el texto. Las cadenas de varias líneas se pueden escribir entre tres comillas.

Print muestra el texto de forma legible. La función repr muestra una representación más precisa para inspección, incluyendo comillas y secuencias de escape. Cuando un texto contiene espacios, saltos o caracteres invisibles que producen un resultado extraño, repr ayuda a detectarlos.

```python
mensaje = "Primera línea\nSegunda línea"
texto_con_espacios = "  aula A-01  \n"

print(mensaje)
print(repr(texto_con_espacios))
```

## 2. Operaciones sobre texto

El operador más concatena cadenas y crea una nueva. El asterisco repite una cadena. Los operadores in y not in preguntan si un fragmento aparece dentro de otro; distinguen mayúsculas y minúsculas. Una comparación de igualdad también distingue mayúsculas y minúsculas.

No confundas concatenar texto con sumar números. El texto 12 y el número 12 son tipos distintos. Cuando construyas un mensaje con un número, conviértelo con str o usa una cadena formateada. No intentes convertir a número un texto que contenga unidades o palabras sin separarlas primero.

```python
aula = "A-01"
prefijo = "A-"
numero_alumnos = 24

print("Aula asignada: " + aula)
print(prefijo in aula)
print("a-" in aula)
print("-")
print(str(numero_alumnos) + " alumnos")
```

## 3. Limpieza y normalización

Los datos de texto suelen tener variaciones irrelevantes: espacios al principio o final, mayúsculas distintas, separadores inconsistentes o saltos de línea. Normalizar no significa destruir información; significa aplicar una regla explícita para que datos equivalentes se puedan comparar o procesar de forma coherente. Conserva el original cuando pueda ser necesario para mostrarlo o auditar una transformación.

Métodos habituales: strip elimina espacios y saltos en los extremos; lower produce minúsculas; upper produce mayúsculas; replace sustituye un fragmento por otro; startswith y endswith comprueban inicio y final; y find devuelve la primera posición de un fragmento o menos uno si no aparece. Los métodos devuelven una nueva cadena. Es recomendable encadenar transformaciones solo cuando cada paso sea comprensible; en una limpieza importante, guarda resultados intermedios con nombres claros.

```python
codigo_original = "  Aula a_01  \n"
sin_extremos = codigo_original.strip()
codigo_normalizado = sin_extremos.lower().replace("_", "-")

print("Original:", repr(codigo_original))
print("Sin extremos:", repr(sin_extremos))
print("Normalizado:", codigo_normalizado)
print(codigo_normalizado.startswith("aula"))
print(codigo_normalizado.find("a-01"))
```

## 4. Separar y unir información

El método split divide una cadena en una lista. Sin argumento, separa por bloques de espacios y trata varios espacios consecutivos como un único separador. Si se indica un separador, como coma o punto y coma, divide exactamente por ese texto. El método join realiza la operación inversa: una cadena separadora une los elementos de una secuencia de textos.

Antes de usar split, conoce el formato esperado. Un dato separado por comas puede contener comas dentro de comillas; en ese caso no basta con split y se necesitará una herramienta específica más adelante. Para ejemplos sencillos con un separador conocido, split y join son adecuados y transparentes.

```python
etiquetas_texto = "energía, aula, consumo"
etiquetas = etiquetas_texto.split(",")
etiquetas_limpias = []

for etiqueta in etiquetas:
    etiquetas_limpias.append(etiqueta.strip().lower())

resumen = " | ".join(etiquetas_limpias)
print(etiquetas)
print(etiquetas_limpias)
print(resumen)
```

## 5. Construir mensajes con claridad

Las cadenas formateadas, también llamadas f-strings, permiten incluir el valor de expresiones entre llaves dentro de un texto. Se escriben colocando la letra f antes de las comillas. Son legibles cuando el mensaje contiene varias variables. Las llaves se reservan para expresiones; si necesitas mostrar una llave literal debes duplicarla.

Una f-string no valida datos ni sustituye a diseñar una salida clara. Decide qué unidades, etiquetas y precisión necesita una persona. Para mostrar un número decimal con dos cifras se puede usar el formato .2f. Mantén separados el cálculo y su presentación: primero calcula un valor numérico; después decide cómo mostrarlo.

```python
aula = "B-02"
temperatura = 21.678
mensaje = f"Aula {aula}: {temperatura:.2f} °C"

print(mensaje)
print(f"La expresión 2 + 3 vale {2 + 3}.")
```

### Preparar texto para un sistema de IA

La limpieza de texto tiene consecuencias. Convertir a minúsculas puede ayudar a tratar Igual y igual como la misma palabra, pero puede eliminar una distinción relevante en códigos o nombres. Eliminar símbolos puede mejorar una búsqueda o destruir unidades, negaciones o información de contexto. No existe una limpieza universal: documenta siempre la regla, conserva si procede el texto original y prueba casos representativos.

Un sistema de IA no corrige automáticamente una preparación de datos deficiente. La calidad de sus resultados depende, entre otras cosas, de que el texto de entrada represente fielmente la información que se quiere analizar.

**Ejercicio 07.02 · Normalizar códigos.** Parte de los códigos [ a_01, A-02, espacio B_03 espacio, b-04 ]. Recorre la lista y construye otra con todos los códigos en mayúsculas, sin espacios externos y usando guiones. Conserva la lista original y muestra ambas.

**Ejercicio 07.03 · Etiquetas de una consulta.** Parte del texto energía, aula , consumo , eficiencia. Divídelo, limpia cada etiqueta, únelas con el separador barra vertical y construye un mensaje formateado que indique cuántas etiquetas se han recibido.


```python
texto = "energía, aula , consumo , eficiencia"

etiquetas = texto.split(",")
etiquetas_limpias = []

for etiqueta in etiquetas:
    etiqueta_limpia = etiqueta.strip()
    etiquetas_limpias.append(etiqueta_limpia)

mensaje_formateado = " | ".join(etiquetas_limpias)
print(f"Se han recibido {len(etiquetas_limpias)} etiquetas: {mensaje_formateado}")
```

**Ejercicio 07.04 · Buscar con intención.** Dado el mensaje La temperatura del aula A-01 es adecuada, comprueba si contiene la palabra temperatura y si termina en adecuada. Explica qué cambios de mayúsculas o puntuación podrían alterar el resultado.


```python
mensaje = "La temperatura del aula A-01 es adecuada"

lista_mensaje = mensaje.split(" ")
contiene_temperatura = None
ultima_palabra = True

for mensaje in lista_mensaje:
    if str(mensaje).lower() == "temperatura":
        contiene_temperatura = True

if str(lista_mensaje[-1]).lower().split() == "adecuada":
    ultima_palabra = True

print(f"Contiene 'temperatura': {contiene_temperatura}")
print(f"Última palabra es 'adecuada': {ultima_palabra}")
```

## Proyecto 07 · Limpieza de comentarios de aula

Construye un programa que prepare estos comentarios para una revisión posterior: comentarios = [ Espacio El aula A-01 está bien Espacio, AULA a_02: revisar ventilación, vacío,  Aula B_01  con buena luz ].

El proyecto debe conservar la lista original y crear una lista de comentarios limpios. Para cada texto debe eliminar espacios externos, convertir a minúsculas y sustituir guiones bajos por guiones. Si un comentario queda vacío después de limpiarlo, debe contarse como vacío y no añadirse a la lista limpia. Debe mostrar un informe numerado de los comentarios conservados, la cantidad de comentarios vacíos y un mensaje formateado final.

Incluye una explicación Markdown de las reglas de limpieza, sus posibles limitaciones y una tabla de pruebas que cubra espacios, mayúsculas, guión bajo y cadena vacía. No apliques transformaciones que no puedas justificar.

```python
comentarios = [ " El aula A-01 está bien ", "AULA a_02: revisar ventilación", "vacío", " Aula B_01 con buena luz " ]
comentarios_limpios = []
cantidad_comentarios_vacios = 0

for comentario in comentarios:
    comentario_limpio = comentario.strip().lower().replace("_", "-")
    if not comentario_limpio:
        cantidad_comentarios_vacios += 1
    else:
        comentarios_limpios.append(comentario_limpio)

for contador, comentario in enumerate(comentarios_limpios, start=1):
    print(f"{contador}. {comentario}")

print(f"Comentarios vacíos: {cantidad_comentarios_vacios}")
print(f"Mensaje formateado: {" | ".join(comentarios)}")
```

| split | strip | replace |
| ---- | ----- | ------- |
| Esta funcion divide la cadena en una lista de palabras | Esta funcion elimina espacios delante y detras de la palabra | Esta funcion sustituye el valor antiguo que eligas por el nuevo |


| Caso de Prueba | Entrada | Salida / Resultado | Comportamiento |
| --- | --- | --- | --- |
| **Espacios** | "   Aula A-01   " | "aula a-01" | Elimina espacios en los extremos y pasa a minúsculas. |
| **Mayúsculas** | "AULA a_02" | "aula a-02" | Normaliza mayúsculas y cambia guion bajo por medio. |
| **Guion bajo** | "B_01 luz" | "b-01 luz" | Sustituye los guiones bajos por guiones estándar. |
| **Vacío** | "" o "   " | *(Omitido)* | Detectado por `if not`, incrementa el contador de vacíos. |

Limitaciones:
- La función `strip()` solo elimina espacios en los extremos, no otros caracteres.
- La función `replace()` sustituye todas las ocurrencias del valor antiguo, no solo la primera.
- La función `lower()` no afecta caracteres que ya están en minúsculas.
