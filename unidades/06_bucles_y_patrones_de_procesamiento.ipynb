## 1. Repetir una acción con for

Un bucle for ejecuta un bloque una vez por cada elemento de una secuencia. Su forma básica tiene una variable de recorrido, la palabra in, la secuencia y dos puntos. En cada repetición, la variable de recorrido se asocia al elemento siguiente. El bloque debe ir sangrado, exactamente igual que los bloques de decisiones.

El nombre de la variable debe describir el elemento individual, no la colección completa. Por ejemplo, para una lista llamada temperaturas, un nombre adecuado es temperatura. Dentro del bucle se trabaja con un elemento; fuera del bucle, la lista continúa disponible. Un for termina automáticamente al llegar al último elemento: no es necesario contar manualmente cuántas repeticiones faltan.

```python
temperaturas = [19.5, 20.1, 21.3]

for temperatura in temperaturas:
    print("Lectura:", temperatura)

print("Fin del recorrido")
```

### Recorrer un rango y conocer la posición

Range representa una sucesión de enteros y permite repetir una operación un número conocido de veces. Recuerda que el final queda excluido. Por ejemplo, range(1, 4) produce 1, 2 y 3. Si necesitas tanto el índice como el elemento de una lista, enumerate entrega ambos valores en cada repetición. Por defecto los índices de enumerate empiezan en 0, pero se puede indicar start=1 para numerar elementos de forma más natural en un informe.

No uses índices si no hacen falta. Recorrer directamente los elementos suele ser más simple y evita IndexError. Usa enumerate cuando la posición aporta significado, por ejemplo al numerar lecturas, o cuando debes modificar una lista por índice de manera controlada.

```python
for numero in range(1, 4):
    print("Intento", numero)

aulas = ["A-01", "A-02", "B-01"]
for posicion, aula in enumerate(aulas, start=1):
    print(posicion, aula)
```

## 2. Patrones de procesamiento

Un bucle es una herramienta; el objetivo es aplicar un patrón de procesamiento. Los tres patrones básicos son:

- **Acumulación:** empezar con un valor inicial y actualizarlo con cada elemento, por ejemplo sumar temperaturas.
- **Conteo:** empezar en cero y aumentar una unidad cuando un elemento cumple una condición.
- **Construcción o filtrado:** empezar con una lista vacía y añadir los elementos o resultados que interesan.

El valor inicial es parte esencial del algoritmo. Una suma empieza en 0 porque cero no altera una suma; un producto empezaría en 1; un contador empieza en 0; y una lista resultado empieza vacía. Inicializar dentro del bucle suele ser un error: reiniciaría el resultado en cada repetición.

```python
lecturas = [19.5, 22.0, 27.5, 21.0]
suma = 0
cantidad_altas = 0
lecturas_adecuadas = []

for lectura in lecturas:
    suma = suma + lectura
    if lectura > 27:
        cantidad_altas = cantidad_altas + 1
    if 18 <= lectura <= 27:
        lecturas_adecuadas.append(lectura)

print("Suma:", suma)
print("Lecturas altas:", cantidad_altas)
print("Lecturas adecuadas:", lecturas_adecuadas)
```

### Cálculos derivados y colecciones vacías

Una media se obtiene dividiendo la suma entre el número de elementos. Antes de dividir hay que saber si existe al menos un elemento: una lista vacía tiene longitud 0 y dividir entre cero genera ZeroDivisionError. En la próxima unidad se verán más herramientas para texto, pero esta misma idea se aplica a cualquier cálculo: identifica las condiciones necesarias para que una operación tenga sentido.

Funciones como sum, min y max pueden resolver algunos casos de forma breve. Debes conocer primero el patrón manual para entender qué cálculo se realiza y poder adaptar la solución cuando haga falta un criterio propio. Min y max también fallan con una secuencia vacía, porque no hay un valor mínimo ni máximo.

```python
lecturas = [19.5, 22.0, 27.5, 21.0]
suma = 0

for lectura in lecturas:
    suma = suma + lectura

if len(lecturas) > 0:
    media = suma / len(lecturas)
    print("Media:", media)
else:
    print("No hay lecturas para calcular una media.")
```

## 3. Repetir mientras se cumpla una condición

Un bucle while ejecuta su bloque mientras una condición sea verdadera. Es apropiado cuando no se conoce de antemano cuántas repeticiones serán necesarias, pero sí se conoce la condición que debe mantenerse. En un while debes preparar tres piezas: un estado inicial, una condición que pueda llegar a ser falsa y una actualización del estado dentro del bloque.

Olvidar la actualización suele causar un bucle infinito. En JupyterLab, si una celda no termina porque un bucle no cambia su condición, utiliza la interrupción del kernel y revisa esas tres piezas. No resuelvas un bucle infinito reiniciando sin entender qué variable debería haberse actualizado.

```python
intento = 1

while intento <= 3:
    print("Comprobación número", intento)
    intento = intento + 1

print("Bucle terminado con intento =", intento)
```

### break y continue: uso prudente

Break termina inmediatamente el bucle más cercano; continue salta el resto de la iteración actual y pasa a la siguiente. Son útiles en situaciones concretas, por ejemplo cuando se localiza un elemento y no hay motivo para seguir buscándolo. Sin embargo, un bucle con muchas salidas puede ser difícil de leer. Antes de usar break o continue, comprueba si una condición clara o una estructura distinta expresaría mejor la intención.

En esta unidad se emplearán solo en ejemplos sencillos. Más importante que memorizar estas palabras es poder explicar cuándo termina un bucle y qué resultado deja cada variable.

```python
lecturas = [20.0, None, 21.5, 28.0, 22.0]

for lectura in lecturas:
    if lectura is None:
        print("Lectura ausente: se omite.")
        continue
    if lectura > 27:
        print("Se encontró una lectura alta:", lectura)
        break
    print("Lectura válida:", lectura)
```

## 4. Depurar bucles

Cuando el resultado de un bucle es incorrecto, no revises solo la última línea. Comprueba: el contenido de la secuencia; los valores iniciales de acumuladores y contadores; qué ocurre en la primera repetición; si cada actualización está dentro o fuera del bloque correcto; y el valor final esperado. Para while, verifica además que la condición puede llegar a ser falsa.

Una traza manual imprime temporalmente el elemento actual y las variables que cambian. El ejemplo siguiente muestra cómo observar la evolución de una suma. Cuando se haya comprobado el algoritmo, retira los mensajes de diagnóstico o sustitúyelos por una salida útil para la persona usuaria.

```python
valores = [2, 5, 3]
total = 0

for valor in valores:
    print("Antes:", valor, total)
    total = total + valor
    print("Después:", valor, total)

print("Total final:", total)
```

## Proyecto 06 · Analizador de lecturas ambientales

Construye un programa que analice estas mediciones de temperatura: lecturas = [17.5, 18.0, 20.4, 27.0, 28.1, None, 22.3]. Una lectura None representa que el sensor no respondió y no debe intervenir en cálculos.

El programa debe crear y mostrar: una lista de lecturas válidas; cantidad de lecturas ausentes; suma y media de las válidas; cantidad de lecturas bajas, adecuadas y altas, usando los límites menor que 18, entre 18 y 27 incluidos, y mayor que 27; y un informe numerado de las lecturas válidas. Debe incluir comentarios sobre los valores iniciales de cada acumulador o contador y una tabla Markdown con casos de prueba que cubra una lectura ausente, los dos límites y valores fuera del rango.

No uses sum, min, max ni comprensiones de listas para resolver el núcleo del proyecto: el objetivo es demostrar que comprendes el recorrido, la acumulación, el conteo y el filtrado. Reinicia el kernel y ejecuta todas las celdas antes de entregar.

```python
lecturas = [17.5, 18.0, 20.4, 27.0, 28.1, None, 22.3]
lecturas_validas = []
lecturas_ausentes = 0
suma_validas = 0
count_bajas = 0
count_adecuadas = 0
count_altas = 0
media_validas = 0

for lectura in lecturas:
    if lectura is None:
        lecturas_ausentes += 1
    else:
        lecturas_validas.append(lectura)
        suma_validas += lectura

        if lectura < 18:
            count_bajas += 1
        elif lectura <= 27:
            count_adecuadas += 1
        else:
            count_altas += 1

total_validas = count_bajas + count_adecuadas + count_altas
if lecturas_validas:
    media_validas = suma_validas / total_validas
else:
    media_validas = 0

print("Informe: ")
for contador, lecturas in enumerate(lecturas_validas, start=1):
    print(f"{contador}. Lectura: {lecturas}")
print("")
print(f"Cantidad de lecturas ausentes: {lecturas_ausentes}")
print(f"Suma de lecturas válidas: {suma_validas}")
print(f"Media de lecturas válidas: {media_validas:.2f}")
print(f"Cantidad de lecturas bajas: {count_bajas}")
print(f"Cantidad de lecturas adecuadas: {count_adecuadas}")
print(f"Cantidad de lecturas altas: {count_altas}")
```

| Caso de Prueba | Valor de Entrada | Razón / Criterio |
|----------------|--------------------|-------------------|
| Lectura ausente | None | El sensor no responde, incrementa lecturas_ausentes |
| Límite inferior | 18.0 | Pertenece al rango $[18, 27]$ por ser exactamente igual a 18 |
| Límite superior | 27.0 | Pertenece al rango $[18, 27]$ por ser exactamente igual a 27 |
| Fuera de rango (Bajo) | 17.5 | Estrictamente menor que 18 ($< 18$) |
| Fuera de rango (Alto) | 28.1 | Estrictamente mayor que 27 ($> 27$) |
