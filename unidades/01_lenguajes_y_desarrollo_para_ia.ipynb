## 1. Qué hace un programa

Un **programa** es un conjunto preciso y ordenado de instrucciones que una máquina puede ejecutar. Normalmente transforma una o varias **entradas** en una **salida** mediante un procesamiento. También puede guardar información para usarla después.

Por ejemplo, un sistema que estima el consumo eléctrico de un aula puede recibir mediciones y horarios (entrada), limpiar y combinar los datos (procesamiento), mostrar una predicción (salida) y registrar los resultados (almacenamiento). Un modelo de IA forma parte del procesamiento: no sustituye a las demás piezas del sistema.

La descomposición entrada–procesamiento–salida ayuda a entender programas antes de conocer todos sus detalles.

````python
# Un ejemplo mínimo en Python.
temperatura = 20       # entrada
aviso = temperatura > 25 # procesamiento
print("¿Conviene ventilar?", aviso)  # salida
````

La respuesta es False dado que el valor de temperatura es menor que 25 y por lo tanto la comprobacion de aviso adquiere un valor boolean de tipo False.

Entrada: 20
Procesamiento: 20 > 25
Salida: False
Guardar: temperatura y aviso

## 2. Del problema a la solución

Programar no consiste en escribir código de inmediato. Un ciclo de desarrollo sencillo incluye estas etapas, que pueden repetirse varias veces:

1. **Comprender el problema:** quién usará el sistema, qué necesita y qué restricciones existen.
2. **Diseñar:** decidir datos, pasos, interfaz, almacenamiento y cómo se comprobará que funciona.
3. **Implementar:** escribir el código y configurar las herramientas necesarias.
4. **Probar:** comprobar casos normales, límites y errores previsibles.
5. **Desplegar y mantener:** poner la solución a disposición de sus usuarios, corregirla y adaptarla.

En IA hay además una cuestión esencial: los datos deben ser pertinentes, estar autorizados y evaluarse para detectar errores, sesgos o resultados inadecuados. Que un modelo produzca una respuesta no demuestra, por sí solo, que la solución sea fiable.

## 3. Elegir un lenguaje de programación

Un lenguaje define la forma de expresar instrucciones. Ningún lenguaje es objetivamente el mejor: la elección depende del problema y del contexto. Conviene comparar, como mínimo, estos criterios:

| Criterio | Pregunta útil |
| --- | --- |
| Legibilidad y aprendizaje | ¿El equipo podrá leer, revisar y mantener el código? |
| Ecosistema | ¿Existen bibliotecas maduras para datos, IA, web o el dispositivo implicado? |
| Rendimiento | ¿Hay requisitos estrictos de tiempo de respuesta, memoria o energía? |
| Integración y despliegue | ¿Con qué sistemas, navegadores, bases de datos o dispositivos debe comunicarse? |
| Comunidad y mantenimiento | ¿Hay documentación, actualizaciones y profesionales disponibles? |
| Seguridad y fiabilidad | ¿Qué controles, revisiones y herramientas ofrece el entorno? |

Python destaca en aprendizaje, automatización y ciencia de datos, y cuenta con un ecosistema amplio para IA. JavaScript/TypeScript suele ser una elección natural para experiencias en el navegador. Java, C# o Kotlin se usan con frecuencia en aplicaciones empresariales y móviles; C/C++ o Rust pueden encajar cuando los recursos o la latencia son críticos. Estas afirmaciones orientan una primera decisión: siempre hay que comprobar las condiciones concretas del proyecto.

### IA: lenguaje, modelo y producto

Una aplicación de IA suele combinar más de un lenguaje. Por ejemplo, se puede entrenar un prototipo en Python, publicar una interfaz con HTML/CSS/JavaScript e integrar el resultado en otro sistema. Elegir Python para el análisis no obliga a que toda la aplicación final esté escrita en Python.

Antes de incorporar IA, formula también: ¿qué decisión apoya?, ¿qué datos utiliza?, ¿quién revisa los resultados?, ¿qué ocurre si se equivoca? Estas preguntas forman parte del diseño técnico y responsable.

## 4. Lenguajes de marcado

Los lenguajes de marcado describen la **estructura** o el significado de información mediante etiquetas. No son, en general, lenguajes destinados a expresar algoritmos como Python.

HTML estructura documentos web. XML permite representar datos con etiquetas definidas para un dominio. En ambos casos, una etiqueta puede contener texto, otras etiquetas o atributos. Observa este ejemplo:

````python
html = """<article id="aviso-1">
  <h1>Consumo del aula</h1>
  <p>La predicción se actualizará mañana.</p>
  <p>Fuente de Datos: Estación del Tiempo</p>
</article>"""
print(html)
````

En el ejemplo, `article`, `h1` y `p` son etiquetas; `id` es un atributo de `article`; y los textos entre las etiquetas son contenido. Las etiquetas de apertura y cierre deben estar correctamente anidadas. En HTML hay algunas excepciones especiales, pero esta regla es una buena base para leer documentos estructurados.
