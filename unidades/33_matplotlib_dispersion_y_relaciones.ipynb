# Unidad 33. Relaciones y diagramas de dispersión con Matplotlib

**Bloque:** Matplotlib (UT3) · **Tiempo orientativo:** 5–7 horas

Un diagrama de dispersión sitúa cada observación mediante dos variables cuantitativas. Es útil para explorar asociación, agrupaciones, valores atípicos y zonas con pocos datos. No demuestra que una variable cause la otra.

**Prerequisitos:** unidades 31 y 32, arrays de NumPy y `DataFrame` de Pandas.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 1. Cada punto representa una observación

En `ax.scatter(x, y)`, la posición horizontal procede de `x` y la vertical de `y`. A diferencia de `plot`, Matplotlib no une automáticamente los puntos: no supone que exista una continuidad ni un orden temporal.

Pregunta siempre qué representa una observación. En el ejemplo, cada punto es una lectura de un aula. La posición sugiere que las lecturas con más personas suelen tener más CO₂, pero no permite estimar una regla general ni identificar otras condiciones que influyen.

```python
personas = np.array([8, 10, 12, 15, 18, 20, 22, 25])
co2_ppm = np.array([520, 560, 590, 650, 720, 710, 790, 830])

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(personas, co2_ppm, color="tab:blue", label="Lectura de aula")
ax.set(title="Ocupación y CO₂ en lecturas puntuales", xlabel="Personas presentes", ylabel="CO₂ (ppm)")
ax.grid(alpha=0.25)
ax.legend()
plt.show()
```

## 2. Color para grupos, no como adorno

Si cada observación pertenece a una categoría, el color puede ayudar a comparar grupos. Para un número reducido de categorías, es preferible dibujar una serie por categoría y dar a cada una una etiqueta en la leyenda. Así no dependes de que la persona lectora adivine qué significa cada color.

No asignes números arbitrarios a categorías y los pases como un gradiente continuo sin explicar la codificación: un mapa continuo sugiere un orden o distancia entre clases que quizá no existe.

```python
datos = pd.DataFrame(
    {
        "personas": [8, 12, 17, 21, 9, 14, 18, 24],
        "co2_ppm": [510, 580, 690, 760, 545, 620, 705, 850],
        "ventilacion": ["Alta", "Alta", "Alta", "Alta", "Baja", "Baja", "Baja", "Baja"],
    }
)

fig, ax = plt.subplots(figsize=(7, 4))
estilos = {"Alta": {"color": "tab:blue", "marker": "o"}, "Baja": {"color": "tab:orange", "marker": "s"}}
for nivel, grupo in datos.groupby("ventilacion"):
    ax.scatter(grupo["personas"], grupo["co2_ppm"], label=f"Ventilación {nivel.lower()}", **estilos[nivel])

ax.set(title="Lecturas según nivel de ventilación", xlabel="Personas presentes", ylabel="CO₂ (ppm)")
ax.grid(alpha=0.25)
ax.legend(title="Condición")
plt.show()
```

## 3. Tamaño y transparencia: más información, más responsabilidad

`s` controla el área de los marcadores, en puntos tipográficos al cuadrado. Si el tamaño codifica una medida, indícalo en el título o la leyenda y elige un rango moderado: círculos enormes pueden ocultar otros puntos y hacer difícil comparar valores.

`alpha` controla la transparencia entre 0 y 1. Con muchos puntos, `alpha` menor que 1 hace visibles las zonas de solapamiento. No cambia los datos; cambia cuánto se aprecia la densidad visual.

```python
tiempo_min = np.array([12, 15, 18, 21, 25, 29, 33])
energia_wh = np.array([30, 38, 43, 54, 61, 70, 79])
muestras = np.array([2, 4, 3, 6, 5, 8, 7])

fig, ax = plt.subplots(figsize=(7, 4))
puntos = ax.scatter(
    tiempo_min,
    energia_wh,
    s=muestras * 45,
    alpha=0.65,
    color="tab:green",
    edgecolor="black",
    label="Tamaño proporcional al número de muestras",
)
ax.set(title="Energía y tiempo de proceso", xlabel="Tiempo (minutos)", ylabel="Energía (Wh)")
ax.grid(alpha=0.25)
ax.legend(loc="upper left")
plt.show()

print("El primer marcador tiene tamaño", puntos.get_sizes()[0], "puntos².")
```

## 4. Asociación no es causalidad

Una nube ascendente indica que, dentro de la muestra, valores altos de `x` coinciden con frecuencia con valores altos de `y`. Puede haber excepciones, variables no observadas o una muestra demasiado pequeña. Evita expresiones como «x provoca y» si el gráfico es la única evidencia disponible.

Una descripción responsable podría ser: «En estas ocho lecturas, el CO₂ aumenta en general al crecer la ocupación; se necesitarían mediciones repetidas y datos sobre ventilación para investigar una posible explicación».

El conjunto siguiente relaciona horas de práctica y resultado de una actividad, con una categoría de turno. Ejecuta el código y después:

### Ejercicio 33.02 · Categorías visibles

Crea un `DataFrame` con ocho observaciones de `x`, `y` y una categoría con los valores `"A"` y `"B"`. Dibuja un grupo por categoría con marcador distinto y leyenda. Comprueba que la leyenda contiene exactamente las dos categorías.

```python
# Escribe aquí tu solución del ejercicio 33.02.
df = pd.DataFrame({
    "x": [1, 2, 3, 4, 1.5, 2.5, 3.5, 4.5],
    "y": [10, 14, 18, 22, 12, 15, 19, 25],
    "categoria": ["A", "A", "A", "A", "B", "B", "B", "B"],
})

# Se extraen los datos que contengan esa categoria.
a = df[df["categoria"]=="A"]
b = df[df["categoria"]=="B"]

fig, ax = plt.subplots()
ax.scatter(a["x"], a["y"], marker="x", label="Categoria A")
ax.scatter(b["x"], b["y"], marker="s", label="Categoria B")

leyenda = ax.legend()
etiquetas = [texto.get_text() for texto in leyenda.get_texts()]
assert etiquetas == ["Categoria A", "Categoria B"], f"Etiquetas inesperadas: {etiquetas}"
print("Etiquetas verificadas:", etiquetas)

plt.show()
```

### Ejercicio 33.03 · Una tercera medida

Con `duracion = [5, 8, 10, 13, 17]`, `energia = [14, 21, 26, 35, 48]` y `repeticiones = [1, 3, 2, 5, 4]`, crea una dispersión cuyo tamaño sea `repeticiones * 50` y cuya transparencia sea `0.6`. Explica qué representa el tamaño y por qué no basta el color para comunicarlo.

```python
# Escribe aquí tu solución del ejercicio 33.03.
duracion = [5, 8, 10, 13, 17]
energia = [14, 21, 26, 35, 48]
repeticiones = [1, 3, 2, 5, 4]

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(
    duracion,
    energia,
    s=np.array(repeticiones) * 50,
    alpha=0.6,
)
ax.set(
    title="Duración y energía según repeticiones",
    xlabel="Duración (minutos)",
    ylabel="Energía (Wh)",
)
plt.show()

# Qué representa el tamaño: el área de cada marcador (s=np.array(repeticiones) * 50) representa el número de repeticiones ([1, 3, 2, 5, 4]), 
# de modo que los puntos más grandes corresponden a más repeticiones y la transparencia (alpha=0.6) ayuda a ver solapamientos.  
# 
# Por qué no basta el color: un color plano o un mapa de color continuo sin leyenda o escala explícita puede confundirse con categorías discretas 
# o sugerir una escala o distancia arbitraria entre valores, mientras que codificarlo en el área (s) 
# y documentarlo en el gráfico comunica de forma más directa la magnitud numérica.  
```

### Ejercicio 33.04 · Función para un grupo

Define `dispersión_por_grupo(datos, columna_x, columna_y, columna_grupo)` que cree y devuelva `fig, ax`. Debe recorrer los grupos, añadir una serie por grupo y crear una leyenda. Pruébala con un `DataFrame` de dos grupos y verifica el número de colecciones de puntos con `len(ax.collections)`.

```python
# Escribe aquí tu solución del ejercicio 33.04.

# columna_grupo es la columna que indica a qué categoría pertenece cada fila, y groupby(columna_grupo) 
# separa las filas según esos valores para que cada grupo se dibuje por separado en el gráfico.
def dispersión_por_grupo(datos, columna_x, columna_y, columna_grupo):
    fig, ax = plt.subplots()
    
    # subgrupo es la tabla con las filas de "A" o de "B"
    # groupby("columna_grupo") separa las filas de tu tabla según el valor que tengan en la columna "grupo", 
    # formando un bloque o grupo por cada valor distinto (en este caso, un grupo para "A" y otro para "B").
    for grupo, subgrupo in datos.groupby(columna_grupo):
        ax.scatter(subgrupo[columna_x], subgrupo[columna_y], label=grupo)
    ax.legend()
    return fig, ax

df_prueba = pd.DataFrame(
    {
        "x": [1, 2, 3, 4, 5, 6],
        "y": [10, 12, 15, 20, 22, 25],
        "grupo": ["A", "A", "A", "B", "B", "B"],
    }
)

fig, ax = dispersión_por_grupo(df_prueba, "x", "y", "grupo")
assert len(ax.collections) == 2
print("Número de colecciones:", len(ax.collections))
plt.show()
```

### Ejercicio 33.05 · Miniproyecto: lecturas de rendimiento

Crea un `DataFrame` con al menos diez observaciones de tiempo de proceso (minutos), consumo energético (Wh), tamaño de lote y tipo de equipo (`"A"` o `"B"`). Elabora una dispersión donde posición represente tiempo y energía, color/marcador represente el equipo y tamaño represente el lote. Añade una conclusión de 70–100 palabras que describa un patrón, una excepción y dos límites del análisis.

```python
# Escribe aquí tu solución del ejercicio 33.05.
df = pd.DataFrame({
    "tiempo": [8, 10, 12, 14, 16, 18, 7, 11, 13, 15, 17, 20],
    "energia": [18, 22, 27, 31, 36, 40, 29, 20, 25, 30, 34, 43],
    "lote": [20, 30, 40, 50, 60, 80, 25, 35, 45, 55, 70, 90],
    "equipo": ["A", "A", "A", "A", "A", "A", "B", "B", "B", "B", "B", "B"],
})

fig, ax = plt.subplots()
# subgrupo es la tabla con las filas de "A" o de "B"
for equipo, subgrupo in df.groupby("equipo"):
    ax.scatter(
        subgrupo["tiempo"],
        subgrupo["energia"],
        # Escalamos el tamaño de lote (* 4) porque `s` en ax.scatter() define el área en pt² y así se distinguen mejor las diferencias entre lotes pequeños y grandes.
        s=np.array(subgrupo["lote"]) * 4,
        alpha=0.6,
        label=equipo,
    )

ax.set(
    title="Rendimiento por equipo y tamaño de lote",
    xlabel="Tiempo (min)",
    ylabel="Energía (Wh)",
)
ax.legend()
plt.show()



```
