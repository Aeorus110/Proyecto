Una función reúne una tarea con nombre, entradas y resultado. Permite evitar repeticiones, probar una parte concreta del programa y expresar una idea de mayor nivel que una secuencia de instrucciones. Esta unidad introduce el diseño de funciones pequeñas, claras y comprobables.

## 1. Por qué usar funciones

Cuando varias partes de un programa realizan la misma tarea, copiar y pegar el código crea problemas: una corrección debe repetirse en todos los lugares, el programa se vuelve largo y es difícil comprobar cada comportamiento. Una función centraliza una tarea bajo un nombre descriptivo.

Una buena función tiene una responsabilidad concreta. Por ejemplo, convertir una lectura de texto a número o clasificar una temperatura son tareas distintas de mostrar un informe. Separarlas permite reutilizarlas y probarlas con datos pequeños. Una función no tiene que ser complicada para ser útil; de hecho, las funciones cortas y cohesionadas suelen ser más fáciles de leer.

## 2. Definir y llamar una función

Una definición comienza con def, seguido de un nombre, paréntesis, dos puntos y un bloque sangrado. La definición registra la función; su cuerpo no se ejecuta hasta que se llama mediante el nombre seguido de paréntesis. Por eso una llamada debe aparecer después de la definición que necesita.

Los parámetros son nombres definidos en la cabecera de la función. Los argumentos son los valores concretos que se proporcionan al llamarla. Cada llamada puede usar argumentos distintos, y la misma función realiza su tarea con ellos.

```python
def mostrar_aula(aula):
    print("Aula:", aula)

mostrar_aula("A-01")
mostrar_aula("B-02")
```

### Devolver resultados

Print muestra información, pero no entrega un resultado que otra parte del programa pueda reutilizar. Return termina la función y devuelve un valor a quien la llamó. Una función que calcula debe normalmente devolver el resultado, mientras que la presentación se puede dejar fuera.

Si una función llega al final sin return, devuelve None. Esto no es necesariamente un error: puede ser adecuado en una función cuyo único propósito es realizar una acción. Sin embargo, si esperabas usar el resultado en un cálculo y obtienes None, revisa si falta un return o si se ha llamado a una función que solo imprime.

```python
def convertir_a_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius

resultado = convertir_a_celsius(68)
print(f"Resultado: {resultado:.1f} °C")
```

## 3. Parámetros, argumentos y valores por defecto

Los argumentos pueden proporcionarse por posición, siguiendo el orden de los parámetros, o por nombre. Los argumentos por nombre aclaran una llamada cuando hay varios valores del mismo tipo. Un parámetro puede tener un valor por defecto, que se utiliza si la llamada no proporciona ese argumento.

Los valores por defecto deben ser seguros. Evita usar una lista o diccionario mutable como valor por defecto, porque se crearía una sola vez y se compartiría entre llamadas. Cuando se necesite una colección opcional, se suele usar None y crear una nueva dentro de la función. Esta regla será especialmente importante al diseñar funciones que acumulen resultados.

```python
def crear_mensaje(aula, estado="adecuada"):
    return f"Aula {aula}: {estado}"

print(crear_mensaje("A-01"))
print(crear_mensaje(aula="B-02", estado="revisar ventilación"))
```

## 4. Alcance y efectos secundarios

Una variable creada dentro de una función tiene alcance local: existe para ese cuerpo de función y no se puede usar directamente fuera. Los parámetros también son locales. Esto protege al resto del programa de nombres temporales y permite reutilizar el mismo nombre en funciones distintas.

Una función puede leer valores globales, pero depender de ellos la hace más difícil de probar y reutilizar. Es preferible recibir por parámetro los datos que necesita y devolver los resultados. Modificar una lista o diccionario recibido como argumento es posible porque son mutables; antes de hacerlo, documenta claramente esa decisión o crea una copia para evitar efectos secundarios inesperados.

```python
temperatura_exterior = 16.0

def diferencia_temperatura(temperatura_interior, temperatura_exterior):
    diferencia = temperatura_interior - temperatura_exterior
    return diferencia

print(diferencia_temperatura(22.0, temperatura_exterior))
```

## 5. Documentación, nombres y anotaciones

El nombre de una función debe expresar una acción o una pregunta, como calcular_media o es_lectura_valida. Una docstring es el primer texto entre comillas dentro del cuerpo y explica qué hace la función, qué espera y qué devuelve. Las anotaciones de tipo indican el tipo esperado de parámetros y retorno; ayudan a lectores, editores y analizadores, pero Python no las aplica automáticamente durante la ejecución.

Una anotación no convierte ni valida datos. Si una función anota que espera un float y recibe texto, seguirá siendo necesario comprobar o convertir el dato. En este curso se utilizan anotaciones sencillas como documentación adicional, sin ocultar la lógica real de validación.

```python
def clasificar_temperatura(temperatura: float) -> str:
    """Devuelve una categoría para una temperatura en grados Celsius."""
    if temperatura < 18:
        return "baja"
    if temperatura > 27:
        return "alta"
    return "adecuada"

print(clasificar_temperatura(17.5))
print(clasificar_temperatura.__doc__)
```

## 6. Comprobar funciones

Una función se prueba llamándola con casos elegidos y comparando el resultado obtenido con el esperado. Assert detiene la ejecución con un mensaje si una condición es falsa. Es útil para comprobaciones pequeñas y reproducibles durante el aprendizaje.

Para una función con límites, incluye casos normales y los valores de frontera. Para una conversión, incluye un valor positivo, cero y uno negativo cuando tenga sentido. Una prueba que solo muestra una salida con print obliga a inspección manual; una aserción deja clara la expectativa y detecta automáticamente una regresión.

```python
assert clasificar_temperatura(17.9) == "baja"
assert clasificar_temperatura(18.0) == "adecuada"
assert clasificar_temperatura(27.0) == "adecuada"
assert clasificar_temperatura(27.1) == "alta"
print("Pruebas superadas.")
```

### Diseñar antes de escribir

Antes de definir una función, responde: qué datos necesita, qué tipo de resultado debe devolver, qué casos límite debe gestionar y si debe modificar alguno de sus argumentos. Una función que hace muchas tareas a la vez suele ser difícil de nombrar y probar. Divide por responsabilidades: una puede limpiar texto, otra clasificar una medida y otra preparar un mensaje.

No uses una función solo para ocultar código. La llamada debe ser más expresiva que el bloque que sustituye, y la función debe conservar una idea coherente.

Ejercicio 10.03 · Evitar efectos secundarios. Define agregar_revision que reciba una lista de tareas y una tarea nueva. Primero crea una versión que modifique la lista original y demuestra el efecto. Después define otra que copie la lista, añada la tarea a la copia y devuelva esa copia. Explica la diferencia.

```python
def agregar_revision(lista_tareas: list, tarea_nueva) -> list:
    lista_tareas.append(tarea_nueva)
    for tarea in lista_tareas:
        print(f"Tarea: {tarea}")

def agregar_revision_2v(lista_tareas: list, tarea_nueva) -> list:
    copia_lista = lista_tareas.copy()
    copia_lista.append(tarea_nueva)
    return copia_lista

#La diferencia que en el primer metodo se edita directamente en la lista original y en el segundo metodo se edita en una copia que no altera la muestra original por si hubiera algun fallo poder tenerla
```

## Proyecto 10 · Biblioteca de funciones para lecturas

Crea tres funciones: convertir_lectura que reciba texto y devuelva float; clasificar_lectura que reciba una temperatura y devuelva baja, adecuada o alta con los límites 18 y 27; y resumir_lecturas que reciba una lista de temperaturas válidas y devuelva un diccionario con cantidad, suma, media y clasificaciones.

Cada función debe tener nombre claro, docstring, anotaciones y una responsabilidad. La función de resumen debe usar las otras cuando resulte pertinente, no repetir su lógica. Crea pruebas assert de las tres funciones, incluyendo los límites, una lista de una lectura y una lista vacía. Para la lista vacía, decide y documenta una política razonable: por ejemplo, media igual a None.

Finalmente, usa las funciones con las entradas de texto [17.5, 18, 22.3, 27, 28.1] y muestra el informe. Reinicia el kernel y ejecuta todo. No uses variables globales para pasar datos entre funciones.

```python
def convertir_lectura(lectura: str) -> float:
    """Convierte una lectura en float"""
    return float(lectura)

def clasificar_lectura(temperatura:float) -> str:
    """Clasifica en funcion de los parametros 18 y 27 el tipo de temperatura"""
    if temperatura < 18:
        return "Baja"
    elif 18 <= temperatura <= 27:
        return "Adecuada"
    else:
        return "Alta"

def resumir_lecturas(lista_temperaturas: list) -> dict:
    """Construye y devuleve un diccionario con los valores del listado"""
    if not lista_temperaturas:
        return "Lista vacia"
    temperaturas_convertidas = []
    lista_clasificaciones = {}
    for temperatura in lista_temperaturas:
        lectura_convertida = convertir_lectura(temperatura)
        temperaturas_convertidas.append(convertir_lectura(lectura_convertida))
        lista_clasificaciones[lectura_convertida] = clasificar_lectura(lectura_convertida)

    diccionario = {
        "cantidad": len(temperaturas_convertidas),
        "suma": sum(temperaturas_convertidas),
        "media": round(sum(temperaturas_convertidas)/len(temperaturas_convertidas), ndigits= 2),
        "clasificaciones": lista_clasificaciones
    }
    return diccionario

assert convertir_lectura("20.5") == 20.5
assert clasificar_lectura(17) == "Baja"
assert clasificar_lectura(22) == "Adecuada"
assert clasificar_lectura(29) == "Alta"
assert resumir_lecturas(["21"])
assert resumir_lecturas([]) #Antes de llegar a que se produzca media/None compruebo si la lista está vacia para acabar el programa
print("---INFORME---")
for lectura in resumir_lecturas(["17.5", "18", "22.3", "27", "28.1"]).items():
    print(lectura)

```
