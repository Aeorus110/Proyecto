```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 1. Histogramas: valores agrupados en intervalos

`ax.hist(valores, bins=...)` reparte los valores numéricos en intervalos (*bins*) y muestra cuántas observaciones caen en cada uno. No representa un valor por barra: cada barra resume un rango. Por ello, el número y anchura de intervalos influyen en la forma que percibimos.

Empieza con un número moderado de intervalos y prueba alternativas. Si la interpretación cambia por completo, indícalo en vez de elegir la versión más llamativa. El histograma no muestra por sí solo qué observación concreta cae en cada intervalo.

```python
tiempos_min = np.array([12, 14, 15, 16, 17, 17, 18, 19, 20, 20, 21, 22, 24, 25, 28])

fig, ax = plt.subplots(figsize=(7, 4))
frecuencias, limites, barras = ax.hist(tiempos_min, bins=5, color="tab:blue", edgecolor="white")
ax.set(title="Distribución de tiempos de proceso", xlabel="Tiempo (minutos)", ylabel="Número de ejecuciones")
ax.grid(axis="y", alpha=0.25)
plt.show()

print("Frecuencias por intervalo:", frecuencias.astype(int))
print("Límites de los intervalos:", np.round(limites, 1))
```

## 2. El número de intervalos es una decisión

Con pocos intervalos se obtiene una visión resumida; con muchos, aparecen detalles y también ruido. No hay un valor universalmente correcto de `bins`: depende del número de observaciones, de las unidades y de la pregunta.

En una comparación de grupos, utiliza los mismos límites e intervalos si quieres que las alturas sean comparables. La transparencia (`alpha`) permite superponer distribuciones, pero puede ser difícil de leer cuando hay muchos grupos. En esos casos, crea paneles separados, como aprenderás en la unidad 35.

```python
turno_manana = np.array([14, 15, 16, 17, 17, 18, 19, 20, 21])
turno_tarde = np.array([16, 18, 19, 20, 21, 22, 23, 24, 26])
limites = np.arange(12, 29, 2)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(turno_manana, bins=limites, alpha=0.6, label="Mañana")
ax.hist(turno_tarde, bins=limites, alpha=0.6, label="Tarde")
ax.set(title="Tiempos de proceso por turno", xlabel="Tiempo (minutos)", ylabel="Número de ejecuciones")
ax.grid(axis="y", alpha=0.25)
ax.legend(title="Turno")
plt.show()
```

## 3. Barras para categorías

`ax.bar(categorias, valores)` representa una cantidad asociada a cada categoría. Antes de dibujar, escribe qué mide la altura: ¿número de casos, porcentaje, media, suma? No son intercambiables.

Ordenar las barras ayuda a comparar magnitudes, excepto cuando existe un orden natural, como meses o niveles definidos. Las barras horizontales (`ax.barh`) resultan útiles si las etiquetas son largas. Añadir el valor encima o junto a la barra mejora la lectura, pero no sustituye el eje y su unidad.

```python
categorias = np.array(["Validado", "Pendiente", "Revisar", "Descartado"])
cantidades = np.array([28, 11, 7, 4])
orden = np.argsort(cantidades)

fig, ax = plt.subplots(figsize=(7, 4))
barras = ax.barh(categorias[orden], cantidades[orden], color="tab:purple")
ax.set(title="Estado de 50 registros", xlabel="Número de registros")
ax.bar_label(barras, padding=3)
ax.grid(axis="x", alpha=0.25)
plt.show()
```

## 4. Barras agrupadas: una comparación adicional

Las barras agrupadas permiten comparar dos medidas de la misma escala dentro de cada categoría. Cada grupo debe tener pocas barras y una leyenda explícita. Con demasiadas series se vuelven ilegibles; una tabla o una figura separada puede ser mejor.

Observa que en este ejemplo las métricas ya están calculadas. Visualizarlas no entrena un modelo ni permite concluir cómo se comportará fuera de estas observaciones.

```python
clases = np.array(["A", "B", "C"])
precision = np.array([0.84, 0.76, 0.88])
cobertura = np.array([0.79, 0.81, 0.72])
posiciones = np.arange(len(clases))
anchura = 0.36

fig, ax = plt.subplots(figsize=(7, 4))
barras_precision = ax.bar(posiciones - anchura / 2, precision, anchura, label="Precisión")
barras_cobertura = ax.bar(posiciones + anchura / 2, cobertura, anchura, label="Cobertura")
ax.set(title="Métricas precomputadas por clase", xlabel="Clase", ylabel="Valor de la métrica")
ax.set_xticks(posiciones, clases)
ax.set_ylim(0, 1)
ax.bar_label(barras_precision, fmt="%.2f", padding=2)
ax.bar_label(barras_cobertura, fmt="%.2f", padding=2)
ax.legend()
ax.grid(axis="y", alpha=0.25)
plt.show()
```

## Práctica guiada

La siguiente tabla contiene tiempos de respuesta y área de atención. Realiza estas acciones:

1. Calcula la media de tiempo por área con `groupby`.
2. Decide si una barra debe representar conteo o media y justifica la elección.
3. Modifica el gráfico para ordenar las áreas por valor.

```python
incidencias = pd.DataFrame(
    {
        "area": ["Red", "Equipo", "Red", "Cuenta", "Equipo", "Red", "Cuenta"],
        "minutos": [18, 32, 21, 14, 28, 25, 17],
    }
)
resumen = incidencias.groupby("area", as_index=False)["minutos"].mean().sort_values("minutos")

fig, ax = plt.subplots()
barras = ax.barh(resumen["area"], resumen["minutos"], color="tab:cyan")
ax.set(title="Tiempo medio de respuesta por área", xlabel="Tiempo medio (minutos)")
ax.bar_label(barras, fmt="%.1f", padding=3)
ax.grid(axis="x", alpha=0.25)
plt.show()
print(resumen)
```

### Ejercicio 34.02 · Misma escala para comparar

Representa dos histogramas de `grupo_a = [12, 13, 14, 15, 15, 16, 17]` y `grupo_b = [15, 16, 17, 18, 18, 19, 20]` con los mismos límites `np.arange(10, 22, 2)`, transparencia y leyenda. ¿Qué detalle del código garantiza que la comparación sea justa?

### Ejercicio 34.03 · Barras ordenadas

Con categorías `['Error', 'Aviso', 'Correcto']` y cantidades `[9, 14, 37]`, crea barras horizontales ordenadas por cantidad. Añade etiquetas numéricas sobre las barras y un eje con la unidad correcta.

```python
# Escribe aquí tu solución del ejercicio 34.03.
categorias = np.array(["Error", "Aviso", "Correcto"])
cantidades = np.array([9, 14, 37])
orden = np.argsort(cantidades) # devuelve los índices (posiciones) que ordenarían el array de menor a mayor, en vez de devolver los valores ordenados directamente.

fig, ax = plt.subplots()
barras = ax.barh(categorias[orden], cantidades[orden])
ax.set(
    title="Registros por estado",
    xlabel="Número de registros",
    ylabel="Estado",
)
ax.bar_label(barras, padding=3)
ax.grid(axis="x", alpha=0.25)
plt.show()
```

### Ejercicio 34.05 · Miniproyecto: distribución y resumen

Crea un `DataFrame` con al menos 20 tiempos de atención (minutos) y una categoría de prioridad (`"Alta"` o `"Normal"`). Entrega dos gráficos: un histograma con intervalos compartidos para ambos grupos y unas barras que comparen la media por prioridad. Añade 80–110 palabras que expliquen qué aporta cada gráfico, qué observas y por qué no basta la media para describir toda la distribución.

```python
# Escribe aquí tu solución del ejercicio 34.05.
df_atencion = pd.DataFrame(
    {
        "minutos": [8, 10, 11, 12, 13, 14, 15, 16, 18, 25, 14, 16, 18, 20, 22, 24, 26, 28, 30, 35],
        "prioridad": ["Alta"] * 10 + ["Normal"] * 10,
    }
)

limites = np.arange(5, 40, 5)

# Gráfico 1: Histograma con intervalos compartidos
fig, ax = plt.subplots()
for prioridad, subgrupo in df_atencion.groupby("prioridad"):
    ax.hist(subgrupo["minutos"], bins=limites, alpha=0.6, edgecolor="white", label=prioridad)
ax.set(
    title="Distribución de tiempos de atención por prioridad",
    xlabel="Tiempo de atención (minutos)",
    ylabel="Número de casos",
)
ax.grid(axis="y", alpha=0.25)
ax.legend(title="Prioridad")
plt.show()

# Gráfico 2: Barras con la media por prioridad
resumen = df_atencion.groupby("prioridad", as_index=False)["minutos"].mean().sort_values("minutos")

fig, ax = plt.subplots()
barras = ax.bar(resumen["prioridad"], resumen["minutos"])
ax.set(
    title="Tiempo medio de atención por prioridad",
    xlabel="Prioridad",
    ylabel="Tiempo medio (minutos)",
)
ax.bar_label(barras, fmt="%.1f", padding=3)
ax.grid(axis="y", alpha=0.25)
plt.show()
```
