## 1. Dos productos diferentes

`A * B` multiplica valores que comparten posición. `A @ B` compone transformaciones: las columnas de `A` deben coincidir con las filas de `B`. Ambas operaciones pueden devolver una matriz de la misma forma, pero significan cosas distintas. Comprueba siempre las formas antes de calcular.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("Elemento a elemento:")
print(a * b)
print("Matricial:")
print(a @ b)
```

## 2. Transpuesta e identidad

La transpuesta intercambia filas y columnas. La identidad es una matriz cuadrada que no cambia otra matriz compatible al multiplicarla. Estas propiedades permiten comprobar resultados, pero una identidad no puede arreglar una incompatibilidad de formas.

```python
c = np.arange(1, 10).reshape(3, 3)
identidad = np.eye(3, dtype=int)
print(c.T)
print(c @ identidad)
```

## 3. Resolver sistemas

Para resolver `A @ x = b`, usa `np.linalg.solve(A, b)` cuando `A` es cuadrada y existe una solución única. Evita calcular una inversa sólo para multiplicarla por `b`: es menos claro y puede ser menos estable. Si el sistema es singular, `solve` debe avisarte y la situación necesita otra formulación.

```python
a = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([5.0, 6.0])
x = np.linalg.solve(a, b)
print(x)
print(a @ x)
```

## 4. Normas y coma flotante

Una norma resume la magnitud de un vector. La norma 2 se relaciona con distancia euclidiana; la norma 1 suma valores absolutos. Los decimales se almacenan con precisión finita: para comprobar cálculos usa `np.allclose`, no igualdad exacta. El número de condición alerta de sistemas sensibles a pequeños cambios.

```python
vector = np.array([3.0, 4.0, 0.0])
print(np.linalg.norm(vector, ord=1))
print(np.linalg.norm(vector, ord=2))
print(np.allclose(0.1 + 0.2, 0.3))
```

**29.01.** Calcula `A * B` y `A @ B` para dos matrices 2×2. Explica la diferencia.

```python
matrizA = np.arange(4).reshape(2,2)
matrizB = np.arange(4,8).reshape(2,2)

# A*B -> producto elemento a elemento (multiplica cada posición por la que ocupa su mismo sitio en la otra matriz
elemento_a_elemento = matrizA*matrizB

#A@B -> producto matricial (multiplica cada fila de matrizA por cada columna de matrizB y suma los productos).
matricial = matrizA @ matrizB

print("Producto elemento a elemento (A * B):\n", elemento_a_elemento)
print("\nProducto matricial (A @ B):\n", matricial)
```

**29.02.** Crea una identidad 4×4 y comprueba que `C @ I` conserva una matriz `C` 4×4.

```python
C = np.arange(1,17).reshape(4,4)

I = np.eye(4, dtype=int)

resultado = I @ C
es_identica = np.array_equal(C, resultado)

print("Identidad I (4x4):\n", I)
print("\nResultado C @ I:\n", resultado)
print("\n¿Conserva exactamente la matriz C?:", es_identica)
```

**29.03.** Resuelve el sistema de tres incógnitas proporcionado y verifica `A @ x` con `np.allclose`.


```python
A = np.array([
    [ 2.0,  1.0, -1.0],
    [-3.0, -1.0,  2.0],
    [-2.0,  1.0,  2.0]
])

b = np.array([8.0, -11.0, -3.0])

x = np.linalg.solve(A,b)
es_valida = np.allclose(A @ x, b)

print("Solución x:", x)                  # [ 2.  3. -1.]
print("Comprobación A @ x:", A @ x)      # [  8. -11.  -3.]
print("¿Coincide con b?:", es_valida)    # True
```

**29.04.** Para dos vectores, calcula normas L1 y L2; indica cuál representa distancia euclidiana.


```python
v1 = np.array([3.0, 4.0, 0.0])
v2 = np.array([1.0, -2.0, 2.0])

# Paso 2: Calcular normas L1 y L2 de cada vector (y de su diferencia)
print("v1 -> Norma L1:", np.linalg.norm(v1, ord=1), "| Norma L2:", np.linalg.norm(v1, ord=2))
print("v2 -> Norma L1:", np.linalg.norm(v2, ord=1), "| Norma L2:", np.linalg.norm(v2, ord=2))

# Distancia entre ambos vectores
diff = v1 - v2
print("\nDistancia L1 (Manhattan):", np.linalg.norm(diff, ord=1))
print("Distancia L2 (Euclidiana):", np.linalg.norm(diff, ord=2))
```

**29.05.** Calcula determinante y número de condición de una matriz invertible y otra singular. Explica por qué una matriz singular no debe invertirse.

```python
M_inv = np.array([[2.0, 1.0], [1.0, 3.0]]) # Matriz invertible (M_inv): filas independientes entre sí
M_sing = np.array([[1.0, 2.0], [2.0, 4.0]]) # Matriz singular (M_sing): una fila es múltiplo exacto de la otra

det_inv = np.linalg.det(M_inv) # Si es distinto de 0, la matriz tiene inversa; si es 0 (o prácticamente 0 por coma flotante), es singular.
det_sing = np.linalg.det(M_sing)
# Calcula el número de condición, que mide cuánto se amplifican los errores numéricos ante pequeñas variaciones. 
# En una matriz bien condicionada es un número pequeño; en una singular tiende a infinito
cond_inv = np.linalg.cond(M_inv)
cond_sing = np.linalg.cond(M_sing)

print(f"Invertible -> Determinante: {det_inv:.2f} | Nº condición: {cond_inv:.2f}")
print(f"Singular   -> Determinante: {det_sing:.2f} | Nº condición: {cond_sing:.2e}")
```

**En la matriz invertible (M_inv):** el determinante es distinto de cero (5.00) y el número de condición es bajo (2.62), lo que indica que el sistema es estable y tiene solución única.   

**En la matriz singular (M_sing):** el determinante es 0.00 (porque la segunda fila es múltiplo exacto de la primera, aportando información redundante) y el número de condición tiende a infinito (un valor del orden de $10^{16}$ por la precisión finita de coma flotante).   

**Por qué no debe invertirse una matriz singular:** matemáticamente, calcular la inversa exige dividir entre el determinante; al ser 0, la división no está definida (provoca un LinAlgError: Singular matrix en np.linalg.inv o solve), y en sistemas casi singulares cualquier error diminuto de redondeo en los datos de entrada se amplifica de forma descontrolada.
