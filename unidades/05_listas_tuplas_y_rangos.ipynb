## 1. Secuencias y listas

Una **secuencia** es una colección ordenada de elementos. El orden importa: la posición permite recuperar cada elemento y puede haber repetidos. Listas, tuplas y rangos son secuencias; las cadenas también lo son, aunque se estudiarán con detalle más adelante.

Una lista se escribe entre corchetes y separa los elementos con comas. Puede almacenar distintos tipos, aunque es preferible que sus elementos tengan el mismo significado. Por ejemplo, una lista de temperaturas es más clara que una mezcla de temperaturas, nombres y mensajes. Una lista vacía se escribe con dos corchetes sin contenido.

Las listas son **mutables**: se pueden modificar después de crearlas. Esto es útil cuando llegan nuevas lecturas o se actualiza una tarea, pero también significa que un cambio puede afectar a otra parte del programa que esté usando la misma lista.

```python
temperaturas = [20.5, 21.0, 22.3, 21.8]
aulas = ["A-01", "A-02", "B-01"]
lista_vacia = []

print(temperaturas)
print(aulas)
print(lista_vacia)
print(type(temperaturas))
```

## 2. Índices y cortes

Cada elemento tiene una posición, llamada **índice**. Python empieza en cero: el primer elemento está en la posición 0, el segundo en la 1 y el último de una secuencia de longitud n está en n menos uno. La función len devuelve esa longitud. También hay índices negativos: menos uno es el último elemento y menos dos el anterior. Si se pide una posición inexistente aparece un IndexError.

Un **corte** recupera una parte de una secuencia. Su forma es secuencia[inicio:fin:paso]. El inicio se incluye y el fin se excluye. Por eso, una lista con índices 1, 2 y 3 se obtiene mediante el corte [1:4]. Si se omite inicio se parte del principio; si se omite fin se llega al final; y el paso determina cada cuántos elementos se toma uno. Un corte normal crea una nueva lista, sin modificar la original.

```python
lecturas = [18.5, 19.0, 20.2, 21.1, 21.7, 22.0]

print("Primero:", lecturas[0])
print("Último:", lecturas[-1])
print("Longitud:", len(lecturas))
print("Posiciones 1 a 3:", lecturas[1:4])
print("Tres primeras:", lecturas[:3])
print("Desde la posición 3:", lecturas[3:])
print("Alternas:", lecturas[::2])
print("Orden inverso:", lecturas[::-1])
```

## 3. Modificar listas y realizar copias

Puedes sustituir un elemento asignando un valor a un índice existente. Entre los métodos más usados están append, que añade al final; insert, que añade en una posición; remove, que borra la primera aparición de un valor; y pop, que extrae y devuelve un elemento, por defecto el último. Estos métodos cambian la lista y normalmente devuelven None. Por ello, no asignes el resultado de append al nombre de la lista: perderías la referencia a la lista.

La asignación no realiza una copia. Si se escribe plan_b = plan_a, los dos nombres se refieren a la misma lista. Una modificación a través de cualquiera de ellos será visible desde el otro. Para una copia independiente de una lista simple usa plan_a.copy() o el corte completo plan_a[:]. La expresión con el operador is permite comprobar si dos nombres se refieren al mismo objeto.

Remove falla si el valor no está presente y pop con una posición inexistente falla por índice. Antes de modificar, comprueba mentalmente el contenido, el tamaño y si realmente deseas cambiar la lista original.

```python
plan_original = ["analizar", "documentar"]
otro_nombre = plan_original
copia_independiente = plan_original.copy()

otro_nombre.append("publicar")
copia_independiente.insert(1, "revisar")
tarea_extraida = copia_independiente.pop()

print("Extraída:", tarea_extraida)
print("Original:", plan_original)
print("Otro nombre:", otro_nombre)
print("Copia:", copia_independiente)
print("¿Original y copia son el mismo objeto?", plan_original is copia_independiente)
```

## 4. Tuplas y rangos

Una **tupla** es una secuencia ordenada e inmutable. Se escribe normalmente con paréntesis. La coma es el elemento decisivo: una tupla de un único valor debe terminar con coma. Las tuplas admiten índices, cortes, len y comprobación de pertenencia, pero no métodos que cambien sus elementos. Son adecuadas para configuraciones fijas o datos que forman una unidad, como unas coordenadas o categorías permitidas.

Range representa una sucesión de enteros. range(fin) comienza en 0 y llega hasta fin sin incluirlo; range(inicio, fin) indica ambos extremos; y range(inicio, fin, paso) añade el avance. El final también queda excluido. Un rango representa los números de forma eficiente; list(range(...)) permite mostrar su contenido durante una comprobación. En la próxima unidad se usará para repetir tareas.

```python
coordenadas_sensor = (37.98, -1.13)
niveles = ("bajo", "medio", "alto")
tupla_un_elemento = ("único",)

print(coordenadas_sensor[0])
print(niveles[-1])
print(tupla_un_elemento)
print(list(range(5)))
print(list(range(2, 7)))
print(list(range(10, 0, -2)))
```

## 5. Operaciones y errores frecuentes

El operador más concatena secuencias del mismo tipo y crea una nueva. El operador asterisco repite una secuencia, por ejemplo una lista de ceros. Los operadores in y not in comprueban si un elemento está presente. Estos recursos son útiles, pero no sustituyen a comprender el contenido y el orden de los datos.

Errores comunes: olvidar que el primer índice es 0; creer que el final de un corte está incluido; modificar una lista mediante un segundo nombre pensando que era una copia; o intentar añadir elementos a una tupla. Si una salida sorprende, muestra la secuencia completa, su longitud y las variables que apuntan a ella.

```python
primer_turno = ["A-01", "A-02"]
segundo_turno = ["B-01"]
turnos = primer_turno + segundo_turno
marcas = [0] * 4

print(turnos)
print(marcas)
print("A-02" in turnos)
print("C-01" not in turnos)
```
