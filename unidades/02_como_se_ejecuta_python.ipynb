## 1. El intérprete

El **código fuente** es el texto que escribimos. Python lo procesa mediante el **intérprete**, el programa instalado en el equipo que lee y ejecuta nuestras instrucciones. Cuando escribes `python archivo.py` en una terminal, el intérprete abre ese archivo y lo ejecuta de arriba abajo.

Python comprueba primero que el texto respete sus reglas básicas. Si encuentra un problema de sintaxis, no empieza la ejecución. Si la sintaxis es correcta, ejecuta las instrucciones en orden; durante esa ejecución pueden aparecer otros errores.

````python
import sys

print("Intérprete utilizado:", sys.executable)
print("Versión de Python:", sys.version.split()[0])
````

## 2. Cuadernos y scripts

Un cuaderno Jupyter (`.ipynb`) contiene celdas de texto, código y sus resultados. Es útil para aprender, experimentar y documentar un análisis. Un **script** es normalmente un archivo de texto con extensión `.py` que contiene código Python. Es útil para automatizar tareas y distribuir programas.

Una celda de código se ejecuta con **Mayús+Intro**. El número que aparece entre corchetes, por ejemplo `[3]`, indica que fue la tercera celda ejecutada desde el último reinicio del kernel. El kernel conserva la memoria entre celdas: por eso ejecutar celdas en un orden distinto puede producir resultados confusos.

Para comprobar que un cuaderno funciona de forma reproducible, utiliza **Kernel → Restart Kernel and Run All Cells**. Si una celda depende de una ejecución anterior no visible, este paso ayuda a descubrirlo.

````python
print("Primera instrucción")
print("Segunda instrucción")
print("El intérprete llega aquí al final")
````

## 3. Los errores son información

Un error no es un fracaso del estudiante: es una descripción de que el programa no puede continuar como se ha escrito. Lee siempre el mensaje completo, empezando por la última línea, que suele indicar el tipo de error.

| Tipo | Cuándo aparece | Ejemplo de causa | Primera acción |
| --- | --- | --- | --- |
| **Sintaxis** (`SyntaxError`) | Antes de ejecutar | Falta un paréntesis o unas comillas. | Revisa la línea señalada y la anterior. |
| **Ejecución** (`NameError`, `TypeError`, etc.) | Mientras se ejecuta | Usar un nombre inexistente u operar datos incompatibles. | Lee el tipo y el nombre de la última línea. |
| **Lógica** | El programa termina, pero el resultado no es el esperado. | Aplicar una fórmula equivocada. | Prueba con casos cuyo resultado conozcas. |

No ejecutes una corrección al azar. Formula una hipótesis, cambia lo mínimo necesario, ejecuta de nuevo y verifica si el resultado tiene sentido.

### Cómo leer una traza

Si ejecutaras `print(nombre_usuario)` sin haber definido ese nombre, Python mostraría una traza que termina con `NameError: name 'nombre_usuario' is not defined`. La última línea nombra el problema; las líneas anteriores localizan dónde ocurrió. En este curso no copies una traza entera sin leerla: resume qué dice y qué cambio vas a comprobar.

## 4. Comentarios, sangría y estilo

Un comentario empieza con `#`. Sirve para explicar una decisión, una intención o un dato no evidente. No debe repetir literalmente una instrucción que ya se entiende.

Python utiliza la **sangría** (espacios al comienzo de una línea) para delimitar bloques. Más adelante se verá al usar decisiones y bucles. Por ahora, respeta estas normas sencillas:

- usa cuatro espacios por nivel cuando haya un bloque; no mezcles tabuladores y espacios;
- elige nombres descriptivos y en minúsculas, con palabras separadas por guiones bajos;
- deja espacios alrededor de operadores, como en `total = precio + iva`;
- usa una instrucción principal por línea;
- escribe código breve y ejecútalo con frecuencia.

Estas convenciones proceden de la guía de estilo PEP 8. No son un adorno: facilitan detectar errores y trabajar con otras personas.

````python
# El valor se muestra para comprobar visualmente el orden de ejecución.
mensaje_de_prueba = "Texto nuevo"
print(mensaje_de_prueba)
````

## 4. Tipos de errores comunes y corrección

## Error de Sintaxis (SyntaxError)
- Código erróneo (no ejecutable):
- print("Hola mundo"

- Mensaje esperado: SyntaxError: unexpected EOF while parsing (indica que falta cerrar un paréntesis o comilla antes de que termine el archivo).

- Versión corregida.

````python
print("Hola mundo")
````

## Error de Nombre (NameError)
- Código erróneo (no ejecutable):
- print(nombre)

- Mensaje esperado: NameError: name 'nombre' is not defined (indica que la variable no ha sido definida antes de usarla).

- Versión corregida.

````python
nombre = "Ana"
print(nombre)
````

## Error de Tipo (TypeError)
- Código erróneo (no ejecutable):
- total = "El resultado es: " + 45

- Mensaje esperado: TypeError: unsupported operand type(s) for +: 'str' and 'int' (indica que no se pueden combinar tipos de datos incompatibles).
- Versión corregida.

````python
total = "El resultado es: " + str(45)
print(total)
````
